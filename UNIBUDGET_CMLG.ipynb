##📘 INSTRUCCIONES DE USO — UNIBUDGET

UniBudget es una aplicación de gestión financiera diseñada para estudiantes universitarios. Permite crear un presupuesto mensual, registrar los gastos realizados durante el mes, consultar cuánto dinero queda disponible, recibir alertas cuando una categoría se acerca a su límite, comparar el plan con lo que realmente ocurrió y guardar los resultados del mes.


✅ Antes de comenzar
Para utilizar correctamente UniBudget:
Ejecuta las celdas en orden, desde la Celda 1 hasta la Celda 7.
No elimines ni cambies las variables del código mientras utilizas la aplicación.
Durante un mismo mes, mantén siempre el mismo nombre del estudiante y el mismo mes cuando registres nuevos gastos.
Antes de iniciar un nuevo mes, guarda primero los resultados utilizando la Celda 6.
Si reinicias completamente la sesión de Google Colab, es posible que los datos guardados temporalmente en memoria se pierdan.




 **Celda 1 — Preparar UniBudget**


Esta es la primera celda que debes ejecutar.
Solo debes presionar el botón Ejecutar.
La celda:
carga las librerías necesarias para que funcione la aplicación;


prepara las herramientas de tablas y gráficas;


y muestra la pantalla de bienvenida de UniBudget.


No debes ingresar ningún dato en esta etapa.
Cuando termine de ejecutarse, continúa con la Celda 2.

**Celda 2 — Crear tu plan financiero**


En esta celda vas a construir el presupuesto que utilizarás durante el mes.

**Paso 1. Ingresa tu mensualidad**

Escribe el valor total de dinero que tienes disponible para el mes.
Por ejemplo:
$2.000.000
UniBudget divide automáticamente este valor siguiendo la regla:
50% Necesidades · 30% Deseos · 20% Ahorro


**Paso 2. Personaliza tus necesidades**


El dinero asignado a Necesidades debe dividirse entre:
Transporte
Alimentación
Materiales universitarios
Debes ingresar un porcentaje para cada categoría.
Los tres porcentajes deben sumar exactamente 100%.
Ejemplo:
Transporte: 20%
Alimentación: 50%
Materiales: 30%
20 + 50 + 30 = 100%
Si los porcentajes no suman 100%, UniBudget mostrará una advertencia y te pedirá ingresarlos nuevamente.


**Paso 3. Revisa tu plan**


Cuando la información sea correcta, la aplicación mostrará:
el presupuesto de cada categoría;
una tabla con la distribución del dinero;
y una gráfica del plan financiero.
Este será el plan base del mes y será utilizado posteriormente para comparar todos tus gastos.
Cuando estés conforme con tu presupuesto, continúa con la Celda 3.


**Celda 3 — Registrar y controlar tus gastos**


Esta celda debe utilizarse cada vez que realices un gasto.
Debes completar:


👤 Estudiante


📅 Mes


🗂️ Categoría


📝 Descripción del gasto


💰 Valor


Ejemplo:
Estudiante: Mariana
Mes: Noviembre
Categoría: Alimentación
Descripción: Almuerzo universidad
Valor: $25.000
Después de completar los campos, ejecuta la celda.


¿Cómo registrar varios gastos?
No debes crear una nueva celda.
Cada vez que tengas un gasto nuevo:


conserva el mismo estudiante;


conserva el mismo mes;


cambia la categoría, descripción y valor;


vuelve a ejecutar la Celda 3.


El gasto nuevo se agrega al historial sin borrar los gastos anteriores.


¿Qué muestra UniBudget después de registrar un gasto?
La aplicación calcula automáticamente:
cuánto dinero tenías presupuestado;
cuánto has gastado hasta el momento;
qué porcentaje del presupuesto has utilizado;
cuánto dinero queda disponible;
y el estado de la categoría.
También muestra alertas según el nivel de consumo del presupuesto.
Por ejemplo:
En control: el gasto todavía está lejos del límite.
Atención: ya utilizaste una parte importante del presupuesto.
Cerca del límite: estás próximo a gastar todo el presupuesto.
Excedido: ya superaste el presupuesto de la categoría.
Importante
Durante un mismo mes debes mantener exactamente el mismo estudiante y mes.
Si cambias alguno de estos datos, UniBudget interpreta que estás comenzando una simulación diferente y crea un historial nuevo.




**Celda 4 — Revisar cómo van tus finanzas**


Puedes ejecutar esta celda en cualquier momento después de haber registrado gastos.
No necesitas ingresar información nueva.
UniBudget toma todos los gastos registrados hasta ese momento y crea un dashboard financiero.
El dashboard muestra:


💰 presupuesto disponible para gastos;


💸 total gastado;


💵 dinero disponible;


💚 ahorro reservado;


🧾 número de gastos registrados;


presupuesto por categoría;
gasto acumulado por categoría;
saldo disponible;
porcentaje utilizado;
y estado de cada categoría.
También genera gráficas como:
Plan vs. gasto actual
Disponible por categoría
Estas gráficas permiten identificar rápidamente qué categorías están bajo control y cuáles necesitan mayor atención.
Importante
Si después de ejecutar esta celda registras un gasto nuevo en la Celda 3, debes volver a ejecutar la Celda 4 para actualizar el dashboard.
Google Colab no actualiza automáticamente los resultados anteriores.

**Celda 5 — Cerrar el mes**


Utiliza esta celda cuando quieras analizar el resultado final del mes.
La aplicación ofrece dos opciones:
Opción 1 — Usar gastos registrados
Selecciona esta opción si durante el mes registraste correctamente tus gastos en la Celda 3.
UniBudget utilizará automáticamente el historial acumulado.
Opción 2 — Ingresar valores finales manualmente
Utiliza esta opción si olvidaste registrar algunos movimientos y prefieres ingresar directamente cuánto gastaste finalmente en cada categoría.
¿Qué calcula la aplicación?
UniBudget compara:
Plan vs. Realidad
y determina:
cuánto se gastó realmente;
cuánto quedó disponible;
si cada categoría estuvo dentro o fuera del presupuesto;
cuál fue el ahorro real;
si existió déficit;
y qué categorías necesitan mejorar.
También genera recomendaciones basadas en los resultados obtenidos.
Ahorro real
El ahorro real se calcula como:
Mensualidad − Gastos reales
Si los gastos son menores que la mensualidad, la diferencia se considera ahorro real.
Si los gastos son mayores que la mensualidad, la aplicación muestra un déficit.

**Celda 6 — Guardar los resultados del mes**


Ejecuta esta celda después de haber realizado el cierre en la Celda 5.
Su objetivo es guardar permanentemente la información del mes.
UniBudget verifica primero que los resultados estén actualizados y después genera un archivo de Excel.
El archivo contiene principalmente:


Historial de gastos


Incluye los movimientos registrados durante el mes:
categoría;
descripción;
valor.


Resumen financiero
Incluye:
presupuesto;
gasto real;
diferencia;
ahorro;
déficit;
y resultados del cierre.


El archivo puede descargarse al computador y utilizarse como registro del mes.

⚠️ Muy importante
Si después de ejecutar la Celda 5 vuelves a la Celda 3 y registras un gasto nuevo:
vuelve a ejecutar la Celda 5;
actualiza el cierre;
y después ejecuta nuevamente la Celda 6.
De esta manera el Excel tendrá la información más reciente.




**Celda 7 — Empezar un nuevo mes**


Esta celda se utiliza únicamente cuando:


ya terminaste el mes actual;


realizaste el cierre;

y guardaste los resultados.


Su función es preparar UniBudget para comenzar un nuevo periodo sin mezclar información con el mes anterior.
Por seguridad, la aplicación solicita dos confirmaciones.


Paso 1
Selecciona:
“Empezar un nuevo mes”


Paso 2
Escribe exactamente:
REINICIAR
en mayúsculas.


Si escribes una palabra diferente, UniBudget cancela el proceso para evitar que elimines accidentalmente la información.
Después de confirmar, se eliminan de la memoria las variables temporales del mes anterior.
Para comenzar nuevamente:


➡️ regresa a la Celda 2


➡️ ingresa la nueva mensualidad


➡️ crea el nuevo presupuesto.

🔄 Resumen del recorrido de UniBudget
El funcionamiento general de la aplicación es:
Celda 1 → Preparar la aplicación

Celda 2 → Crear el presupuesto


Celda 3 → Registrar gastos durante el mes


Celda 4 → Consultar el estado de las finanzas


Celda 5 → Cerrar y analizar el mes

Celda 7 → Preparar un nuevo mes

```python
#@title 💰 CREA TU PLAN FINANCIERO { display-mode: "form" }


# ---------------------------------------------------------
# 1. DEFINICIÓN DE NUESTROS PORCENTAJES 50/30/20
# ---------------------------------------------------------

porcentaje_necesidades = 0.50
porcentajes_deseos = 0.30
porcentaje_ahorro = 0.20


# ---------------------------------------------------------
# 2. INGRESAR MENSUALIDAD
# ---------------------------------------------------------

mensualidad = float(input("Ingresa tu mensualidad: "))


# ---------------------------------------------------------
# 3. DIVIDIR LA MENSUALIDAD CON LA REGLA 50/30/20
# ---------------------------------------------------------

necesidades = mensualidad * porcentaje_necesidades
deseos = mensualidad * porcentajes_deseos
ahorro = mensualidad * porcentaje_ahorro


# ---------------------------------------------------------
# 4. MOSTRAR PRESUPUESTO PRINCIPAL DE FORMA VISUAL
# ---------------------------------------------------------

display(HTML(f"""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 20px;
">

    <div style="
        font-size: 22px;
        font-weight: bold;
        color: #1F4E79;
        margin-bottom: 12px;
    ">
        ✨ Tu presupuesto mensual
    </div>


    <div style="
        display: flex;
        gap: 12px;
        flex-wrap: wrap;
    ">


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #2a78d6;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                🏠 Necesidades
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #1F4E79;
                margin-top: 4px;
            ">
                ${necesidades:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                50% de tu mensualidad
            </div>

        </div>


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #eb6834;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                ✨ Deseos
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #eb6834;
                margin-top: 4px;
            ">
                ${deseos:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                30% de tu mensualidad
            </div>

        </div>


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #1baf7a;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                💚 Ahorro
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #1baf7a;
                margin-top: 4px;
            ">
                ${ahorro:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                20% separado desde el inicio
            </div>

        </div>

    </div>

</div>
"""))


# ---------------------------------------------------------
# 5. DIVISIÓN DEL 50% DE NECESIDADES
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 25px;
    margin-bottom: 10px;
">

    <div style="
        font-size: 20px;
        font-weight: bold;
        color: #1F4E79;
    ">
        🎓 Personaliza tus necesidades
    </div>

    <div style="
        font-size: 13px;
        color: #52514e;
        margin-top: 5px;
    ">
        Decide cómo quieres distribuir el presupuesto de Necesidades.
        Transporte, Alimentación y Materiales deben sumar 100%.
    </div>

</div>
"""))


suma_porcentajes = 0


while suma_porcentajes != 100:

    porcentaje_transporte = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Transporte? ")
    )

    porcentaje_alimentacion = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Alimentación? ")
    )

    porcentaje_materiales = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Materiales? ")
    )


    suma_porcentajes = (
        porcentaje_transporte
        + porcentaje_alimentacion
        + porcentaje_materiales
    )


    # -----------------------------------------------------
    # SI NO SUMAN 100%, MOSTRAR ERROR Y VOLVER A PREGUNTAR
    # -----------------------------------------------------

    if suma_porcentajes != 100:

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            background-color: #fff3f3;
            border-left: 6px solid #d9534f;
            border-radius: 8px;
            padding: 15px;
            margin-top: 12px;
            margin-bottom: 12px;
            color: #8a1f1f;
        ">

            <div style="
                font-size: 18px;
                font-weight: bold;
                margin-bottom: 6px;
            ">
                ⚠️ Revisa tus porcentajes
            </div>

            <div style="
                font-size: 14px;
            ">
                Los porcentajes de Transporte, Alimentación
                y Materiales deben sumar <b>100%</b>.
                <br><br>

                Actualmente suman
                <b>{suma_porcentajes}%</b>.
                <br>

                Por favor, vuelve a intentarlo.
            </div>

        </div>
        """))


# ---------------------------------------------------------
# 6. CALCULAR TRANSPORTE, ALIMENTACIÓN Y MATERIALES
# ---------------------------------------------------------

transporte = necesidades * (porcentaje_transporte / 100)

alimentacion = necesidades * (porcentaje_alimentacion / 100)

materiales = necesidades * (porcentaje_materiales / 100)


# ---------------------------------------------------------
# 7. MOSTRAR DISTRIBUCIÓN DE NECESIDADES
# ---------------------------------------------------------

display(HTML(f"""
<div style="
    font-family: Arial, sans-serif;
    background: #eef4fc;
    border-left: 6px solid #2a78d6;
    border-radius: 8px;
    padding: 16px;
    margin-top: 15px;
">

    <div style="
        font-size: 19px;
        font-weight: bold;
        color: #1F4E79;
        margin-bottom: 10px;
    ">
        🎓 Distribución de tus necesidades
    </div>


    <div style="
        font-size: 14px;
        line-height: 2;
        color: #303030;
    ">

        🚍 Transporte:
        <b>${transporte:,.0f}</b>
        ({porcentaje_transporte:.0f}% de Necesidades)

        <br>

        🍴 Alimentación:
        <b>${alimentacion:,.0f}</b>
        ({porcentaje_alimentacion:.0f}% de Necesidades)

        <br>

        📚 Materiales:
        <b>${materiales:,.0f}</b>
        ({porcentaje_materiales:.0f}% de Necesidades)

    </div>

</div>
"""))


# ---------------------------------------------------------
# 8. ORGANIZAR LOS VALORES EN UNA TABLA DE PANDAS
# ---------------------------------------------------------

categorias = [
    "Transporte",
    "Alimentación",
    "Materiales",
    "Deseos",
    "Ahorro"
]


presupuestos = [
    transporte,
    alimentacion,
    materiales,
    deseos,
    ahorro
]


presupuesto = pd.DataFrame({

    "Categoría": categorias,

    "Presupuesto": presupuestos

})


# ---------------------------------------------------------
# 9. MOSTRAR LA TABLA DE PANDAS DE FORMA BONITA
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 25px;
    margin-bottom: 8px;
">

    <div style="
        font-size: 21px;
        font-weight: bold;
        color: #1F4E79;
    ">
        📋 Tu plan financiero del mes
    </div>

    <div style="
        font-size: 13px;
        color: #52514e;
        margin-top: 4px;
    ">
        Este es el presupuesto que intentaremos cumplir durante el mes.
    </div>

</div>
"""))


display(

    presupuesto.style

    .format({
        "Presupuesto": "${:,.0f}"
    })

    .background_gradient(
        subset=["Presupuesto"],
        cmap="Blues"
    )

    .set_caption(
        "Distribución recomendada de tu mensualidad"
    )

)


# ---------------------------------------------------------
# 10. GRÁFICA DEL PLAN FINANCIERO
# ---------------------------------------------------------

colores = [

    "#2a78d6",   # Transporte

    "#5b9bd5",   # Alimentación

    "#9dc3e6",   # Materiales

    "#eb6834",   # Deseos

    "#1baf7a"    # Ahorro

]


plt.figure(figsize=(10, 5))


barras = plt.bar(

    presupuesto["Categoría"],

    presupuesto["Presupuesto"],

    color=colores

)


plt.title(
    "Mi plan financiero del mes",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel("Categorías")

plt.ylabel("Presupuesto asignado")


# Mostrar el valor encima de cada barra

plt.bar_label(

    barras,

    labels=[
        f"${valor:,.0f}"
        for valor in presupuesto["Presupuesto"]
    ],

    padding=3

)


# Quitar líneas innecesarias de la gráfica

plt.gca().spines["top"].set_visible(False)

plt.gca().spines["right"].set_visible(False)


plt.tight_layout()

plt.show()


# ---------------------------------------------------------
# 11. MENSAJE FINAL DE LA PRIMERA ETAPA
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    background: #eef8f4;
    border-left: 6px solid #1baf7a;
    border-radius: 8px;
    padding: 16px;
    margin-top: 15px;
    color: #184f3d;
">

    <div style="
        font-size: 18px;
        font-weight: bold;
        margin-bottom: 6px;
    ">
        ✅ Tu plan financiero está listo
    </div>

    <div style="
        font-size: 14px;
    ">
        Este será tu punto de partida para el mes.
        <br><br>

        A continuación podrás registrar tus gastos,
        revisar cuánto presupuesto te queda y recibir
        alertas cuando te acerques al límite de una categoría.
    </div>

</div>
"""))
```

```python
#@title 6️⃣ GUARDA TU MES FINANCIERO { display-mode: "form" }

#@markdown ### 💾 Guarda toda la información de tu mes
#@markdown Se creará un archivo Excel con:
#@markdown - **Historial de gastos**
#@markdown - **Resumen financiero final**


# ---------------------------------------------------------
# 1. IMPORTAR HERRAMIENTAS NECESARIAS
# ---------------------------------------------------------

from google.colab import files as colab_files

from IPython.display import (
    display,
    HTML,
    FileLink
)

from openpyxl import load_workbook

from openpyxl.styles import (
    Font,
    PatternFill,
    Alignment
)


# ---------------------------------------------------------
# 2. COMENZAR VALIDACIÓN
# ---------------------------------------------------------

exportacion_valida = True


# ---------------------------------------------------------
# 3. VERIFICAR QUE EXISTA EL PLAN FINANCIERO
# ---------------------------------------------------------

variables_plan = [
    "mensualidad",
    "transporte",
    "alimentacion",
    "materiales",
    "deseos",
    "ahorro"
]


for variable in variables_plan:

    if variable not in globals():

        exportacion_valida = False


if not exportacion_valida:

    display(HTML("""
    <div style="
        font-family:Arial,sans-serif;
        background:#fff3f3;
        border-left:6px solid #d9534f;
        border-radius:8px;
        padding:16px;
        margin-top:10px;
        color:#8a1f1f;
    ">

        <div style="
            font-size:20px;
            font-weight:bold;
            margin-bottom:7px;
        ">
            ⚠️ No existe un plan financiero completo
        </div>

        Ejecuta primero la <b>Celda 2</b>
        para crear tu presupuesto.

    </div>
    """))


# ---------------------------------------------------------
# 4. VERIFICAR QUE EXISTA ESTUDIANTE Y MES
# ---------------------------------------------------------

if exportacion_valida:

    if (
        "estudiante" not in globals()
        or "mes" not in globals()
    ):

        exportacion_valida = False

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fff3f3;
            border-left:6px solid #d9534f;
            border-radius:8px;
            padding:16px;
            margin-top:10px;
            color:#8a1f1f;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                ⚠️ No existe un seguimiento activo
            </div>

            Ejecuta primero la <b>Celda 3</b>
            para identificar al estudiante y el mes.

        </div>
        """))


# ---------------------------------------------------------
# 5. VERIFICAR QUE SE HAYA CERRADO EL MES EN CELDA 5
# ---------------------------------------------------------

if exportacion_valida:

    if (
        "modo_cierre" not in globals()
        or "resultado_final" not in globals()
    ):

        exportacion_valida = False

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:16px;
            margin-top:10px;
            color:#765700;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                🏁 Primero debes cerrar el mes
            </div>

            Ejecuta la <b>Celda 5</b>
            antes de guardar tu información.

            <br><br>

            Así nos aseguramos de guardar exactamente
            el cierre financiero que acabas de revisar.

        </div>
        """))


# ---------------------------------------------------------
# 6. OBTENER LOS VALORES REALES SEGÚN EL MODO DE CIERRE
# ---------------------------------------------------------

if exportacion_valida:


    # =====================================================
    # OPCIÓN A:
    # USAR LOS GASTOS REGISTRADOS
    # =====================================================

    if modo_cierre == "Usar gastos registrados":


        if (
            "gastos" not in globals()
            or len(gastos) == 0
        ):

            exportacion_valida = False

            display(HTML("""
            <div style="
                font-family:Arial,sans-serif;
                background:#fffbea;
                border-left:6px solid #e0a800;
                border-radius:8px;
                padding:16px;
                color:#765700;
            ">

                ⚠️ Seleccionaste
                <b>Usar gastos registrados</b>,
                pero actualmente no hay gastos registrados.

                <br><br>

                Revisa la <b>Celda 3</b>
                y vuelve a ejecutar la <b>Celda 5</b>.

            </div>
            """))


        else:


            # Crear tabla con los gastos actuales

            gastos_exportacion_df = pd.DataFrame(

                gastos,

                columns=[
                    "Categoría",
                    "Descripción",
                    "Valor"
                ]

            )


            # Sumar gastos por categoría

            totales_exportacion = (

                gastos_exportacion_df

                .groupby("Categoría")["Valor"]

                .sum()

            )


            transporte_exportar = totales_exportacion.get(
                "Transporte",
                0
            )


            alimentacion_exportar = totales_exportacion.get(
                "Alimentación",
                0
            )


            materiales_exportar = totales_exportacion.get(
                "Materiales",
                0
            )


            deseos_exportar = totales_exportacion.get(
                "Deseos",
                0
            )


    # =====================================================
    # OPCIÓN B:
    # USAR LOS VALORES FINALES MANUALES
    # =====================================================

    else:


        variables_manuales = [

            "transporte_real_manual",

            "alimentacion_real_manual",

            "materiales_real_manual",

            "deseos_real_manual"

        ]


        manuales_disponibles = True


        for variable in variables_manuales:

            if variable not in globals():

                manuales_disponibles = False


        if not manuales_disponibles:

            exportacion_valida = False

            display(HTML("""
            <div style="
                font-family:Arial,sans-serif;
                background:#fffbea;
                border-left:6px solid #e0a800;
                border-radius:8px;
                padding:16px;
                color:#765700;
            ">

                ⚠️ No encontramos los valores manuales
                utilizados para cerrar el mes.

                <br><br>

                Ejecuta nuevamente la <b>Celda 5</b>
                antes de guardar.

            </div>
            """))


        else:

            transporte_exportar = (
                transporte_real_manual
            )

            alimentacion_exportar = (
                alimentacion_real_manual
            )

            materiales_exportar = (
                materiales_real_manual
            )

            deseos_exportar = (
                deseos_real_manual
            )


# ---------------------------------------------------------
# 7. CALCULAR RESULTADOS FINALES
# ---------------------------------------------------------

if exportacion_valida:


    total_gastos_exportar = (

        transporte_exportar

        + alimentacion_exportar

        + materiales_exportar

        + deseos_exportar

    )


    # -----------------------------------------------------
    # CALCULAR AHORRO REAL
    # -----------------------------------------------------

    saldo_exportar = (

        mensualidad

        - total_gastos_exportar

    )


    if saldo_exportar >= 0:

        ahorro_exportar = (
            saldo_exportar
        )

        deficit_exportar = 0


    else:

        ahorro_exportar = 0

        deficit_exportar = abs(
            saldo_exportar
        )


    # -----------------------------------------------------
    # PORCENTAJE REAL DE AHORRO
    # -----------------------------------------------------

    if mensualidad > 0:

        porcentaje_ahorro_exportar = (

            ahorro_exportar
            / mensualidad

        ) * 100


    else:

        porcentaje_ahorro_exportar = 0


# ---------------------------------------------------------
# 8. CREAR PLAN Y REALIDAD PARA COMPROBAR CELDA 5
# ---------------------------------------------------------

if exportacion_valida:


    categorias_exportar = [

        "Transporte",

        "Alimentación",

        "Materiales",

        "Deseos",

        "Ahorro"

    ]


    plan_exportar = [

        transporte,

        alimentacion,

        materiales,

        deseos,

        ahorro

    ]


    realidad_exportar = [

        transporte_exportar,

        alimentacion_exportar,

        materiales_exportar,

        deseos_exportar,

        ahorro_exportar

    ]


# ---------------------------------------------------------
# 9. COMPROBAR QUE LA CELDA 5 ESTÉ ACTUALIZADA
# ---------------------------------------------------------

if exportacion_valida:


    cierre_coincide = True


    try:


        for i in range(
            len(categorias_exportar)
        ):


            categoria_actual = (
                categorias_exportar[i]
            )


            valor_esperado = (
                realidad_exportar[i]
            )


            fila_resultado = resultado_final[

                resultado_final["Categoría"]
                == categoria_actual

            ]


            valor_celda5 = float(

                fila_resultado[
                    "Real"
                ].iloc[0]

            )


            if abs(
                valor_celda5
                - valor_esperado
            ) > 0.01:

                cierre_coincide = False


    except:

        cierre_coincide = False


    # -----------------------------------------------------
    # SI NO COINCIDE, NO GUARDAR
    # -----------------------------------------------------

    if not cierre_coincide:

        exportacion_valida = False

        display(HTML(f"""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:18px;
            margin-top:12px;
            color:#765700;
        ">

            <div style="
                font-size:21px;
                font-weight:bold;
                margin-bottom:8px;
            ">
                🔄 Tu cierre necesita actualizarse
            </div>

            <div style="
                font-size:14px;
                line-height:1.8;
            ">

                Detectamos que los datos actuales de

                <b>{estudiante}</b>

                para

                <b>{mes}</b>

                no coinciden con el último cierre
                realizado en la Celda 5.

                <br><br>

                Esto puede ocurrir si registraste o cambiaste
                gastos después de cerrar el mes.

                <br><br>

                👉 Ejecuta nuevamente la
                <b>Celda 5</b>

                y después vuelve a ejecutar esta
                <b>Celda 6</b>.

                <br><br>

                Así evitamos guardar información incorrecta.

            </div>

        </div>
        """))


# ---------------------------------------------------------
# 10. SI TODO COINCIDE, CREAR EL EXCEL
# ---------------------------------------------------------

if exportacion_valida:


    # =====================================================
    # CREAR HISTORIAL DE GASTOS
    # =====================================================

    if (
        "gastos" in globals()
        and len(gastos) > 0
    ):


        historial_exportar = pd.DataFrame(

            gastos,

            columns=[
                "Categoría",
                "Descripción",
                "Valor"
            ]

        )


    else:


        historial_exportar = pd.DataFrame(

            columns=[
                "Categoría",
                "Descripción",
                "Valor"
            ]

        )


    # Agregar estudiante y mes

    historial_exportar.insert(

        0,

        "Estudiante",

        estudiante

    )


    historial_exportar.insert(

        1,

        "Mes",

        mes

    )


    # -----------------------------------------------------
    # 11. CALCULAR DIFERENCIAS
    # -----------------------------------------------------

    diferencias_exportar = []


    for i in range(
        len(categorias_exportar)
    ):


        diferencia = (

            realidad_exportar[i]

            - plan_exportar[i]

        )


        diferencias_exportar.append(
            diferencia
        )


    # -----------------------------------------------------
    # 12. CALCULAR PORCENTAJE Y RESULTADO
    # -----------------------------------------------------

    porcentajes_exportar = []

    resultados_exportar = []


    for i in range(
        len(categorias_exportar)
    ):


        categoria_actual = (
            categorias_exportar[i]
        )


        plan_actual = (
            plan_exportar[i]
        )


        real_actual = (
            realidad_exportar[i]
        )


        # -------------------------------------------------
        # PORCENTAJE
        # -------------------------------------------------

        if plan_actual > 0:

            porcentaje = (

                real_actual
                / plan_actual

            )


        else:

            if real_actual > 0:

                porcentaje = 1

            else:

                porcentaje = 0


        porcentajes_exportar.append(
            porcentaje
        )


        # -------------------------------------------------
        # AHORRO
        # -------------------------------------------------

        if categoria_actual == "Ahorro":


            if real_actual >= plan_actual:

                resultado = (
                    "Meta alcanzada"
                )


            elif real_actual >= plan_actual * 0.85:

                resultado = (
                    "Cerca de la meta"
                )


            else:

                resultado = (
                    "Meta no alcanzada"
                )


        # -------------------------------------------------
        # GASTOS
        # -------------------------------------------------

        else:


            if real_actual <= plan_actual:

                resultado = (
                    "En presupuesto"
                )


            elif real_actual <= plan_actual * 1.10:

                resultado = (
                    "Exceso leve"
                )


            else:

                resultado = (
                    "Presupuesto excedido"
                )


        resultados_exportar.append(
            resultado
        )


    # -----------------------------------------------------
    # 13. CREAR RESUMEN FINANCIERO
    # -----------------------------------------------------

    resumen_exportar = pd.DataFrame({

        "Categoría":
        categorias_exportar,

        "Plan":
        plan_exportar,

        "Real":
        realidad_exportar,

        "Diferencia":
        diferencias_exportar,

        "% del plan":
        porcentajes_exportar,

        "Resultado":
        resultados_exportar

    })


    # -----------------------------------------------------
    # 14. CREAR NOMBRE DEL ARCHIVO
    # -----------------------------------------------------

    nombre_estudiante = (

        estudiante
        .strip()
        .replace(" ", "_")

    )


    nombre_mes = (

        mes
        .strip()
        .replace(" ", "_")

    )


    nombre_excel = (

        "Finanzas_"

        + nombre_estudiante

        + "_"

        + nombre_mes

        + ".xlsx"

    )


    # -----------------------------------------------------
    # 15. CREAR ARCHIVO EXCEL
    # -----------------------------------------------------

    with pd.ExcelWriter(

        nombre_excel,

        engine="openpyxl"

    ) as writer:


        # HOJA 1

        historial_exportar.to_excel(

            writer,

            sheet_name="Historial de gastos",

            index=False,

            startrow=4

        )


        # HOJA 2

        resumen_exportar.to_excel(

            writer,

            sheet_name="Resumen financiero",

            index=False,

            startrow=10

        )


    # -----------------------------------------------------
    # 16. ABRIR EXCEL PARA DARLE FORMATO
    # -----------------------------------------------------

    libro = load_workbook(
        nombre_excel
    )


    # Colores

    azul = "2A78D6"

    azul_oscuro = "1F4E79"

    naranja = "EB6834"

    verde = "1BAF7A"

    rojo = "D9534F"

    amarillo = "E0A800"

    gris_claro = "F5F7FA"

    gris = "52514E"

    blanco = "FFFFFF"


    # =====================================================
    # HOJA 1 — HISTORIAL DE GASTOS
    # =====================================================

    hoja_gastos = libro[
        "Historial de gastos"
    ]


    # -----------------------------------------------------
    # TÍTULO
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A1:E1"
    )


    hoja_gastos["A1"] = (
        "Historial de gastos"
    )


    hoja_gastos["A1"].font = Font(

        size=18,

        bold=True,

        color=blanco

    )


    hoja_gastos["A1"].fill = PatternFill(

        "solid",

        fgColor=azul_oscuro

    )


    hoja_gastos["A1"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # ESTUDIANTE Y MES
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A2:E2"
    )


    hoja_gastos["A2"] = (

        estudiante

        + " | "

        + mes

    )


    hoja_gastos["A2"].font = Font(

        italic=True,

        color=gris

    )


    hoja_gastos["A2"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # NOTA SOBRE EL MODO DE CIERRE
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A3:E3"
    )


    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):

        hoja_gastos["A3"] = (

            "Nota: el resumen final usa los valores "
            "manuales ingresados en el cierre. "
            "Este historial muestra únicamente los "
            "gastos registrados en la aplicación."

        )


    else:

        hoja_gastos["A3"] = (

            "El resumen financiero fue calculado "
            "con los gastos registrados en este historial."

        )


    hoja_gastos["A3"].font = Font(

        italic=True,

        size=10,

        color=gris

    )


    hoja_gastos["A3"].alignment = Alignment(

        horizontal="center",

        wrap_text=True

    )


    # -----------------------------------------------------
    # ENCABEZADOS DE TABLA
    # -----------------------------------------------------

    for celda in hoja_gastos[5]:


        celda.font = Font(

            bold=True,

            color=blanco

        )


        celda.fill = PatternFill(

            "solid",

            fgColor=azul

        )


        celda.alignment = Alignment(

            horizontal="center"

        )


    # -----------------------------------------------------
    # FORMATO DE MONEDA
    # -----------------------------------------------------

    for fila in range(

        6,

        hoja_gastos.max_row + 1

    ):


        hoja_gastos[

            f"E{fila}"

        ].number_format = '$#,##0'


    # Congelar encabezados

    hoja_gastos.freeze_panes = "A6"


    # Filtro

    hoja_gastos.auto_filter.ref = (

        f"A5:E{hoja_gastos.max_row}"

    )


    # Ancho de columnas

    hoja_gastos.column_dimensions["A"].width = 18

    hoja_gastos.column_dimensions["B"].width = 15

    hoja_gastos.column_dimensions["C"].width = 20

    hoja_gastos.column_dimensions["D"].width = 42

    hoja_gastos.column_dimensions["E"].width = 18


    # =====================================================
    # HOJA 2 — RESUMEN FINANCIERO
    # =====================================================

    hoja_resumen = libro[
        "Resumen financiero"
    ]


    # -----------------------------------------------------
    # TÍTULO
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A1:F1"
    )


    hoja_resumen["A1"] = (
        "Resumen financiero del mes"
    )


    hoja_resumen["A1"].font = Font(

        size=18,

        bold=True,

        color=blanco

    )


    hoja_resumen["A1"].fill = PatternFill(

        "solid",

        fgColor=azul_oscuro

    )


    hoja_resumen["A1"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # ESTUDIANTE Y MES
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A2:F2"
    )


    hoja_resumen["A2"] = (

        estudiante

        + " | "

        + mes

    )


    hoja_resumen["A2"].font = Font(

        italic=True,

        color=gris

    )


    hoja_resumen["A2"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # INDICADORES GENERALES
    # -----------------------------------------------------

    hoja_resumen["A4"] = (
        "Mensualidad"
    )

    hoja_resumen["B4"] = (
        mensualidad
    )


    hoja_resumen["C4"] = (
        "Total gastado"
    )

    hoja_resumen["D4"] = (
        total_gastos_exportar
    )


    hoja_resumen["A5"] = (
        "Ahorro real"
    )

    hoja_resumen["B5"] = (
        ahorro_exportar
    )


    hoja_resumen["C5"] = (
        "% ahorro"
    )

    hoja_resumen["D5"] = (

        porcentaje_ahorro_exportar
        / 100

    )


    hoja_resumen["A6"] = (
        "Meta de ahorro"
    )

    hoja_resumen["B6"] = (
        ahorro
    )


    hoja_resumen["C6"] = (
        "Déficit"
    )

    hoja_resumen["D6"] = (
        deficit_exportar
    )


    hoja_resumen["A7"] = (
        "Modo de cierre"
    )

    hoja_resumen["B7"] = (
        modo_cierre
    )


    # -----------------------------------------------------
    # NOTA DEL CIERRE
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A8:F8"
    )


    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):

        hoja_resumen["A8"] = (

            "El resumen utiliza los valores finales "
            "ingresados manualmente en la Celda 5."

        )


    else:

        hoja_resumen["A8"] = (

            "El resumen utiliza los gastos registrados "
            "durante el seguimiento."

        )


    hoja_resumen["A8"].font = Font(

        italic=True,

        size=10,

        color=gris

    )


    hoja_resumen["A8"].alignment = Alignment(

        horizontal="center",

        wrap_text=True

    )


    # -----------------------------------------------------
    # FORMATO DE LOS INDICADORES
    # -----------------------------------------------------

    for celda in [

        "A4",
        "C4",
        "A5",
        "C5",
        "A6",
        "C6",
        "A7"

    ]:


        hoja_resumen[
            celda
        ].font = Font(

            bold=True,

            color=azul_oscuro

        )


        hoja_resumen[
            celda
        ].fill = PatternFill(

            "solid",

            fgColor=gris_claro

        )


    # Moneda

    for celda in [

        "B4",
        "D4",
        "B5",
        "B6",
        "D6"

    ]:


        hoja_resumen[
            celda
        ].number_format = '$#,##0'


    # Porcentaje

    hoja_resumen[
        "D5"
    ].number_format = "0%"


    # -----------------------------------------------------
    # ENCABEZADOS PLAN VS. REALIDAD
    # -----------------------------------------------------

    for celda in hoja_resumen[11]:


        celda.font = Font(

            bold=True,

            color=blanco

        )


        celda.fill = PatternFill(

            "solid",

            fgColor=azul

        )


        celda.alignment = Alignment(

            horizontal="center"

        )


    # -----------------------------------------------------
    # FORMATEAR FILAS DEL RESUMEN
    # -----------------------------------------------------

    for fila in range(

        12,

        hoja_resumen.max_row + 1

    ):


        # Moneda

        hoja_resumen[
            f"B{fila}"
        ].number_format = '$#,##0'


        hoja_resumen[
            f"C{fila}"
        ].number_format = '$#,##0'


        hoja_resumen[
            f"D{fila}"
        ].number_format = '$#,##0'


        # Porcentaje

        hoja_resumen[
            f"E{fila}"
        ].number_format = "0%"


        # Resultado

        resultado = hoja_resumen[
            f"F{fila}"
        ].value


        if resultado in [

            "En presupuesto",

            "Meta alcanzada"

        ]:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="E8F5EE"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=verde,

                bold=True

            )


        elif resultado in [

            "Cerca de la meta",

            "Exceso leve"

        ]:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="FFF4E8"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=naranja,

                bold=True

            )


        else:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="FFF3F3"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=rojo,

                bold=True

            )


    # Congelar tabla

    hoja_resumen.freeze_panes = "A12"


    # Filtro

    hoja_resumen.auto_filter.ref = (

        f"A11:F{hoja_resumen.max_row}"

    )


    # Anchos de columnas

    hoja_resumen.column_dimensions["A"].width = 22

    hoja_resumen.column_dimensions["B"].width = 22

    hoja_resumen.column_dimensions["C"].width = 18

    hoja_resumen.column_dimensions["D"].width = 18

    hoja_resumen.column_dimensions["E"].width = 16

    hoja_resumen.column_dimensions["F"].width = 25


    # -----------------------------------------------------
    # 17. GUARDAR EXCEL FINAL
    # -----------------------------------------------------

    libro.save(
        nombre_excel
    )


    # -----------------------------------------------------
    # 18. MOSTRAR CONFIRMACIÓN
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        margin-top:10px;
        margin-bottom:18px;
    ">

        <div style="
            font-size:28px;
            font-weight:bold;
            color:#1F4E79;
        ">
            💾 Tu mes financiero está listo
        </div>


        <div style="
            font-size:15px;
            color:#303030;
            margin-top:8px;
        ">

            👤 <b>{estudiante}</b>

            &nbsp;&nbsp; | &nbsp;&nbsp;

            📅 <b>{mes}</b>

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 19. MOSTRAR DATOS QUE SE GUARDARON
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        display:flex;
        gap:12px;
        flex-wrap:wrap;
        margin-bottom:20px;
    ">


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #2a78d6;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                🧾 Gastos registrados
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1F4E79;
            ">
                {len(historial_exportar)}
            </div>

        </div>


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #eb6834;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💸 Gasto final
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#eb6834;
            ">
                ${total_gastos_exportar:,.0f}
            </div>

        </div>


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #1baf7a;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💚 Ahorro real
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1baf7a;
            ">
                ${ahorro_exportar:,.0f}
            </div>

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 20. MOSTRAR QUÉ CONTIENE EL ARCHIVO
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        background:#eef8f4;
        border-left:6px solid #1baf7a;
        border-radius:8px;
        padding:18px;
        margin-top:15px;
        color:#184f3d;
    ">

        <div style="
            font-size:21px;
            font-weight:bold;
            margin-bottom:8px;
        ">
            ✅ Archivo creado correctamente
        </div>

        <div style="
            font-size:14px;
            line-height:1.8;
        ">

            Tu archivo se llama:

            <br>

            📊 <b>{nombre_excel}</b>

            <br><br>

            <b>Hoja 1 — Historial de gastos</b>

            <br>

            Contiene todos los movimientos que
            registraste en la Celda 3.

            <br><br>

            <b>Hoja 2 — Resumen financiero</b>

            <br>

            Contiene el resultado final utilizado
            en tu cierre del mes:

            <br>

            • Mensualidad

            <br>

            • Gastos finales

            <br>

            • Ahorro real

            <br>

            • Déficit

            <br>

            • Plan vs. Realidad

            <br>

            • Resultado de cada categoría

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 21. SI EL CIERRE FUE MANUAL, EXPLICAR LA DIFERENCIA
    # -----------------------------------------------------

    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):


        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:16px;
            margin-top:15px;
            color:#765700;
        ">

            <div style="
                font-size:18px;
                font-weight:bold;
                margin-bottom:6px;
            ">
                ℹ️ Importante sobre tu cierre manual
            </div>

            La hoja <b>Historial de gastos</b>
            muestra únicamente lo que registraste
            durante el mes en la aplicación.

            <br><br>

            La hoja <b>Resumen financiero</b>
            utiliza los valores finales manuales
            que confirmaste en la Celda 5.

            <br><br>

            Por eso ambos valores pueden ser diferentes,
            y eso es correcto.

        </div>
        """))


    # -----------------------------------------------------
    # 22. MOSTRAR ENLACE DE DESCARGA
    # -----------------------------------------------------

    display(HTML("""
    <div style="
        font-family:Arial,sans-serif;
        margin-top:20px;
        font-size:19px;
        font-weight:bold;
        color:#1F4E79;
    ">
        📥 Descarga tu archivo Excel:
    </div>
    """))


    display(

        FileLink(
            nombre_excel
        )

    )


    # -----------------------------------------------------
    # 23. INTENTAR DESCARGA AUTOMÁTICA
    # -----------------------------------------------------

    try:

        colab_files.download(
            nombre_excel
        )


    except:

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:15px;
            margin-top:15px;
            color:#765700;
        ">

            ⚠️ El navegador no inició la descarga automáticamente.

            <br><br>

            Haz clic en el enlace que aparece arriba
            para descargar el archivo manualmente.

        </div>
        """))
```
