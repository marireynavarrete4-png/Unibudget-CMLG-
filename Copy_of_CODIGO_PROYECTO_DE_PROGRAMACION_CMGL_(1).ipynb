```python
#@title 💰 CREA TU PLAN FINANCIERO { display-mode: "form" }


# ---------------------------------------------------------
# 1. DEFINICIÓN DE NUESTROS PORCENTAJES 50/30/20
# ---------------------------------------------------------

porcentaje_necesidades = 0.50
porcentajes_deseos = 0.30
porcentaje_ahorro = 0.20


# ---------------------------------------------------------
# 2. INGRESAR MENSUALIDAD
# ---------------------------------------------------------

mensualidad = float(input("Ingresa tu mensualidad: "))


# ---------------------------------------------------------
# 3. DIVIDIR LA MENSUALIDAD CON LA REGLA 50/30/20
# ---------------------------------------------------------

necesidades = mensualidad * porcentaje_necesidades
deseos = mensualidad * porcentajes_deseos
ahorro = mensualidad * porcentaje_ahorro


# ---------------------------------------------------------
# 4. MOSTRAR PRESUPUESTO PRINCIPAL DE FORMA VISUAL
# ---------------------------------------------------------

display(HTML(f"""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 20px;
">

    <div style="
        font-size: 22px;
        font-weight: bold;
        color: #1F4E79;
        margin-bottom: 12px;
    ">
        ✨ Tu presupuesto mensual
    </div>


    <div style="
        display: flex;
        gap: 12px;
        flex-wrap: wrap;
    ">


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #2a78d6;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                🏠 Necesidades
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #1F4E79;
                margin-top: 4px;
            ">
                ${necesidades:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                50% de tu mensualidad
            </div>

        </div>


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #eb6834;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                ✨ Deseos
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #eb6834;
                margin-top: 4px;
            ">
                ${deseos:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                30% de tu mensualidad
            </div>

        </div>


        <div style="
            flex: 1;
            min-width: 180px;
            background: #f5f7fa;
            border-top: 5px solid #1baf7a;
            border-radius: 8px;
            padding: 14px;
        ">

            <div style="
                font-size: 13px;
                color: #52514e;
            ">
                💚 Ahorro
            </div>

            <div style="
                font-size: 27px;
                font-weight: bold;
                color: #1baf7a;
                margin-top: 4px;
            ">
                ${ahorro:,.0f}
            </div>

            <div style="
                font-size: 12px;
                color: #52514e;
                margin-top: 4px;
            ">
                20% separado desde el inicio
            </div>

        </div>

    </div>

</div>
"""))


# ---------------------------------------------------------
# 5. DIVISIÓN DEL 50% DE NECESIDADES
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 25px;
    margin-bottom: 10px;
">

    <div style="
        font-size: 20px;
        font-weight: bold;
        color: #1F4E79;
    ">
        🎓 Personaliza tus necesidades
    </div>

    <div style="
        font-size: 13px;
        color: #52514e;
        margin-top: 5px;
    ">
        Decide cómo quieres distribuir el presupuesto de Necesidades.
        Transporte, Alimentación y Materiales deben sumar 100%.
    </div>

</div>
"""))


suma_porcentajes = 0


while suma_porcentajes != 100:

    porcentaje_transporte = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Transporte? ")
    )

    porcentaje_alimentacion = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Alimentación? ")
    )

    porcentaje_materiales = float(
        input("¿Qué porcentaje de Necesidades quieres destinar a Materiales? ")
    )


    suma_porcentajes = (
        porcentaje_transporte
        + porcentaje_alimentacion
        + porcentaje_materiales
    )


    # -----------------------------------------------------
    # SI NO SUMAN 100%, MOSTRAR ERROR Y VOLVER A PREGUNTAR
    # -----------------------------------------------------

    if suma_porcentajes != 100:

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            background-color: #fff3f3;
            border-left: 6px solid #d9534f;
            border-radius: 8px;
            padding: 15px;
            margin-top: 12px;
            margin-bottom: 12px;
            color: #8a1f1f;
        ">

            <div style="
                font-size: 18px;
                font-weight: bold;
                margin-bottom: 6px;
            ">
                ⚠️ Revisa tus porcentajes
            </div>

            <div style="
                font-size: 14px;
            ">
                Los porcentajes de Transporte, Alimentación
                y Materiales deben sumar <b>100%</b>.
                <br><br>

                Actualmente suman
                <b>{suma_porcentajes}%</b>.
                <br>

                Por favor, vuelve a intentarlo.
            </div>

        </div>
        """))


# ---------------------------------------------------------
# 6. CALCULAR TRANSPORTE, ALIMENTACIÓN Y MATERIALES
# ---------------------------------------------------------

transporte = necesidades * (porcentaje_transporte / 100)

alimentacion = necesidades * (porcentaje_alimentacion / 100)

materiales = necesidades * (porcentaje_materiales / 100)


# ---------------------------------------------------------
# 7. MOSTRAR DISTRIBUCIÓN DE NECESIDADES
# ---------------------------------------------------------

display(HTML(f"""
<div style="
    font-family: Arial, sans-serif;
    background: #eef4fc;
    border-left: 6px solid #2a78d6;
    border-radius: 8px;
    padding: 16px;
    margin-top: 15px;
">

    <div style="
        font-size: 19px;
        font-weight: bold;
        color: #1F4E79;
        margin-bottom: 10px;
    ">
        🎓 Distribución de tus necesidades
    </div>


    <div style="
        font-size: 14px;
        line-height: 2;
        color: #303030;
    ">

        🚍 Transporte:
        <b>${transporte:,.0f}</b>
        ({porcentaje_transporte:.0f}% de Necesidades)

        <br>

        🍴 Alimentación:
        <b>${alimentacion:,.0f}</b>
        ({porcentaje_alimentacion:.0f}% de Necesidades)

        <br>

        📚 Materiales:
        <b>${materiales:,.0f}</b>
        ({porcentaje_materiales:.0f}% de Necesidades)

    </div>

</div>
"""))


# ---------------------------------------------------------
# 8. ORGANIZAR LOS VALORES EN UNA TABLA DE PANDAS
# ---------------------------------------------------------

categorias = [
    "Transporte",
    "Alimentación",
    "Materiales",
    "Deseos",
    "Ahorro"
]


presupuestos = [
    transporte,
    alimentacion,
    materiales,
    deseos,
    ahorro
]


presupuesto = pd.DataFrame({

    "Categoría": categorias,

    "Presupuesto": presupuestos

})


# ---------------------------------------------------------
# 9. MOSTRAR LA TABLA DE PANDAS DE FORMA BONITA
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    margin-top: 25px;
    margin-bottom: 8px;
">

    <div style="
        font-size: 21px;
        font-weight: bold;
        color: #1F4E79;
    ">
        📋 Tu plan financiero del mes
    </div>

    <div style="
        font-size: 13px;
        color: #52514e;
        margin-top: 4px;
    ">
        Este es el presupuesto que intentaremos cumplir durante el mes.
    </div>

</div>
"""))


display(

    presupuesto.style

    .format({
        "Presupuesto": "${:,.0f}"
    })

    .background_gradient(
        subset=["Presupuesto"],
        cmap="Blues"
    )

    .set_caption(
        "Distribución recomendada de tu mensualidad"
    )

)


# ---------------------------------------------------------
# 10. GRÁFICA DEL PLAN FINANCIERO
# ---------------------------------------------------------

colores = [

    "#2a78d6",   # Transporte

    "#5b9bd5",   # Alimentación

    "#9dc3e6",   # Materiales

    "#eb6834",   # Deseos

    "#1baf7a"    # Ahorro

]


plt.figure(figsize=(10, 5))


barras = plt.bar(

    presupuesto["Categoría"],

    presupuesto["Presupuesto"],

    color=colores

)


plt.title(
    "Mi plan financiero del mes",
    fontsize=16,
    fontweight="bold"
)


plt.xlabel("Categorías")

plt.ylabel("Presupuesto asignado")


# Mostrar el valor encima de cada barra

plt.bar_label(

    barras,

    labels=[
        f"${valor:,.0f}"
        for valor in presupuesto["Presupuesto"]
    ],

    padding=3

)


# Quitar líneas innecesarias de la gráfica

plt.gca().spines["top"].set_visible(False)

plt.gca().spines["right"].set_visible(False)


plt.tight_layout()

plt.show()


# ---------------------------------------------------------
# 11. MENSAJE FINAL DE LA PRIMERA ETAPA
# ---------------------------------------------------------

display(HTML("""
<div style="
    font-family: Arial, sans-serif;
    background: #eef8f4;
    border-left: 6px solid #1baf7a;
    border-radius: 8px;
    padding: 16px;
    margin-top: 15px;
    color: #184f3d;
">

    <div style="
        font-size: 18px;
        font-weight: bold;
        margin-bottom: 6px;
    ">
        ✅ Tu plan financiero está listo
    </div>

    <div style="
        font-size: 14px;
    ">
        Este será tu punto de partida para el mes.
        <br><br>

        A continuación podrás registrar tus gastos,
        revisar cuánto presupuesto te queda y recibir
        alertas cuando te acerques al límite de una categoría.
    </div>

</div>
"""))
```

```python
#@title 3️⃣ REGISTRA Y CONTROLA TUS GASTOS { display-mode: "form" }

#@markdown ### 👤 Identifica tu seguimiento financiero

estudiante = "Isabella De La Concha" #@param {type:"string"}

mes = "Mayo" #@param ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"]

#@markdown ### 🎯 ¿Qué quieres hacer?

accion = "Registrar gasto" #@param ["Registrar gasto", "Reiniciar historial"]

#@markdown ### 📝 Datos del gasto
#@markdown Si seleccionaste **Registrar gasto**, completa los siguientes campos.

categoria = "Alimentación" #@param ["Transporte", "Alimentación", "Materiales", "Deseos"]

descripcion = "pasta" #@param {type:"string"}

valor = 23000 #@param {type:"number"}


# ---------------------------------------------------------
# 1. IDENTIFICAR LA SIMULACIÓN ACTUAL
# ---------------------------------------------------------

simulacion = estudiante.strip() + " - " + mes


# ---------------------------------------------------------
# 2. DETECTAR SI ES UNA NUEVA SIMULACIÓN
# ---------------------------------------------------------

nueva_simulacion = False


# Si nunca se había creado una simulación,
# empezamos desde cero

if "simulacion_actual" not in globals():

    simulacion_actual = simulacion

    gastos = []

    nueva_simulacion = True


# Si cambió el estudiante o el mes,
# comenzamos automáticamente un nuevo historial

elif simulacion_actual != simulacion:

    simulacion_actual = simulacion

    gastos = []

    nueva_simulacion = True


# ---------------------------------------------------------
# 3. MOSTRAR MENSAJE SI COMENZÓ UNA NUEVA SIMULACIÓN
# ---------------------------------------------------------

if nueva_simulacion:

    display(HTML(f"""
    <div style="
        font-family: Arial, sans-serif;
        background: #eef8f4;
        border-left: 6px solid #1baf7a;
        border-radius: 8px;
        padding: 16px;
        margin-top: 12px;
        margin-bottom: 15px;
        color: #184f3d;
    ">

        <div style="
            font-size: 20px;
            font-weight: bold;
            margin-bottom: 7px;
        ">
            🌟 Nuevo seguimiento iniciado
        </div>

        <div style="
            font-size: 14px;
            line-height: 1.7;
        ">

            👤 Estudiante:
            <b>{estudiante}</b>

            <br>

            📅 Mes:
            <b>{mes}</b>

            <br><br>

            El historial de gastos comenzó automáticamente
            desde cero.

        </div>

    </div>
    """))


# ---------------------------------------------------------
# 4. REINICIAR EL HISTORIAL MANUALMENTE
# ---------------------------------------------------------

if accion == "Reiniciar historial":

    gastos = []

    display(HTML(f"""
    <div style="
        font-family: Arial, sans-serif;
        background: #eef8f4;
        border-left: 6px solid #1baf7a;
        border-radius: 8px;
        padding: 18px;
        margin-top: 12px;
        color: #184f3d;
    ">

        <div style="
            font-size: 21px;
            font-weight: bold;
            margin-bottom: 7px;
        ">
            ✅ Historial reiniciado
        </div>

        <div style="
            font-size: 14px;
            line-height: 1.8;
        ">

            El historial financiero de

            <b>{estudiante}</b>

            para

            <b>{mes}</b>

            fue eliminado.

            <br><br>

            Ya puedes comenzar una nueva simulación desde cero.

            <br><br>

            Cambia la opción de arriba a

            <b>Registrar gasto</b>

            y vuelve a ejecutar esta celda.

        </div>

    </div>
    """))


# ---------------------------------------------------------
# 5. SI EL USUARIO QUIERE REGISTRAR UN GASTO
# ---------------------------------------------------------

else:

    # Protección adicional:
    # si por alguna razón la lista no existe, crearla

    if "gastos" not in globals():

        gastos = []


    # -----------------------------------------------------
    # 6. GUARDAR LOS PRESUPUESTOS DE CADA CATEGORÍA
    # -----------------------------------------------------

    limites = {

        "Transporte": transporte,

        "Alimentación": alimentacion,

        "Materiales": materiales,

        "Deseos": deseos

    }


    # -----------------------------------------------------
    # 7. VALIDAR LOS DATOS DEL GASTO
    # -----------------------------------------------------

    datos_validos = True


    # Revisar que el nombre del estudiante no esté vacío

    if estudiante.strip() == "":

        datos_validos = False

        display(HTML("""
        <div style="
            font-family: Arial, sans-serif;
            background: #fff3f3;
            border-left: 6px solid #d9534f;
            border-radius: 8px;
            padding: 15px;
            margin-top: 12px;
            color: #8a1f1f;
        ">

            <b>⚠️ Falta el nombre del estudiante</b>

            <br><br>

            Escribe un nombre antes de registrar el gasto.

        </div>
        """))


    # Revisar que la descripción no esté vacía

    elif descripcion.strip() == "":

        datos_validos = False

        display(HTML("""
        <div style="
            font-family: Arial, sans-serif;
            background: #fff3f3;
            border-left: 6px solid #d9534f;
            border-radius: 8px;
            padding: 15px;
            margin-top: 12px;
            color: #8a1f1f;
        ">

            <b>⚠️ Falta la descripción</b>

            <br><br>

            Escribe una breve descripción del gasto.

        </div>
        """))


    # Revisar que el valor sea mayor que cero

    elif valor <= 0:

        datos_validos = False

        display(HTML("""
        <div style="
            font-family: Arial, sans-serif;
            background: #fff3f3;
            border-left: 6px solid #d9534f;
            border-radius: 8px;
            padding: 15px;
            margin-top: 12px;
            color: #8a1f1f;
        ">

            <b>⚠️ Valor no válido</b>

            <br><br>

            El gasto debe ser mayor que cero.

        </div>
        """))


    # -----------------------------------------------------
    # 8. SI LOS DATOS SON CORRECTOS, GUARDAR EL GASTO
    # -----------------------------------------------------

    if datos_validos:

        gastos.append([

            categoria,

            descripcion.strip(),

            valor

        ])


        # -------------------------------------------------
        # 9. CONVERTIR LOS GASTOS EN UNA TABLA DE PANDAS
        # -------------------------------------------------

        gastos_df = pd.DataFrame(

            gastos,

            columns=[

                "Categoría",

                "Descripción",

                "Valor"

            ]

        )


        # -------------------------------------------------
        # 10. CALCULAR LO GASTADO EN LA CATEGORÍA ACTUAL
        # -------------------------------------------------

        gastado_categoria = gastos_df[

            gastos_df["Categoría"] == categoria

        ]["Valor"].sum()


        # Presupuesto de esa categoría

        limite_categoria = limites[categoria]


        # Dinero disponible

        restante = (

            limite_categoria
            - gastado_categoria

        )


        # -------------------------------------------------
        # 11. CALCULAR PORCENTAJE UTILIZADO
        # -------------------------------------------------

        if limite_categoria > 0:

            porcentaje_usado = (

                gastado_categoria
                / limite_categoria

            ) * 100


        else:

            if gastado_categoria > 0:

                porcentaje_usado = 100

            else:

                porcentaje_usado = 0


        # -------------------------------------------------
        # 12. MOSTRAR CONFIRMACIÓN DEL GASTO
        # -------------------------------------------------

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            background: #eef4fc;
            border-left: 6px solid #2a78d6;
            border-radius: 8px;
            padding: 16px;
            margin-top: 15px;
        ">

            <div style="
                font-size: 20px;
                font-weight: bold;
                color: #1F4E79;
                margin-bottom: 8px;
            ">
                ✅ Gasto registrado correctamente
            </div>

            <div style="
                font-size: 14px;
                color: #303030;
                line-height: 1.8;
            ">

                👤 Estudiante:
                <b>{estudiante}</b>

                <br>

                📅 Mes:
                <b>{mes}</b>

                <br>

                🏷️ Categoría:
                <b>{categoria}</b>

                <br>

                📝 Descripción:
                <b>{descripcion}</b>

                <br>

                💸 Valor:
                <b>${valor:,.0f}</b>

            </div>

        </div>
        """))


        # -------------------------------------------------
        # 13. DEFINIR LA ALERTA FINANCIERA
        # -------------------------------------------------

        if porcentaje_usado >= 100:

            color_fondo = "#fff3f3"

            color_borde = "#d9534f"

            color_texto = "#8a1f1f"

            mensaje = (
                "🚨 Has alcanzado o superado "
                "el presupuesto de esta categoría."
            )


        elif porcentaje_usado >= 85:

            color_fondo = "#fff4e8"

            color_borde = "#eb6834"

            color_texto = "#9a431f"

            mensaje = (
                "⚠️ Estás muy cerca de alcanzar "
                "el límite de esta categoría."
            )


        elif porcentaje_usado >= 70:

            color_fondo = "#fffbea"

            color_borde = "#e0a800"

            color_texto = "#765700"

            mensaje = (
                "🔔 Ya utilizaste más del 70% "
                "de tu presupuesto."
            )


        else:

            color_fondo = "#eef8f4"

            color_borde = "#1baf7a"

            color_texto = "#184f3d"

            mensaje = (
                "✅ Vas bien. Tus gastos siguen "
                "dentro del plan."
            )


        # -------------------------------------------------
        # 14. MOSTRAR ESTADO DE LA CATEGORÍA
        # -------------------------------------------------

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            background: {color_fondo};
            border-left: 6px solid {color_borde};
            border-radius: 8px;
            padding: 16px;
            margin-top: 12px;
            color: {color_texto};
        ">

            <div style="
                font-size: 20px;
                font-weight: bold;
                margin-bottom: 8px;
            ">
                📊 Estado de {categoria}
            </div>

            <div style="
                font-size: 14px;
                line-height: 1.9;
            ">

                {mensaje}

                <br><br>

                💰 Presupuesto:
                <b>${limite_categoria:,.0f}</b>

                <br>

                💸 Gastado hasta ahora:
                <b>${gastado_categoria:,.0f}</b>

                <br>

                📈 Porcentaje utilizado:
                <b>{porcentaje_usado:.0f}%</b>

                <br>

                💵 Saldo disponible:
                <b>${restante:,.0f}</b>

            </div>

        </div>
        """))


        # -------------------------------------------------
        # 15. CREAR BARRA VISUAL DE PROGRESO
        # -------------------------------------------------

        porcentaje_barra = min(

            porcentaje_usado,

            100

        )


        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            margin-top: 16px;
            margin-bottom: 20px;
        ">

            <div style="
                font-size: 14px;
                color: #52514e;
                margin-bottom: 6px;
            ">

                Uso del presupuesto de

                <b>{categoria}</b>

            </div>

            <div style="
                width: 100%;
                background: #e5e7eb;
                border-radius: 10px;
                height: 20px;
                overflow: hidden;
            ">

                <div style="
                    width: {porcentaje_barra}%;
                    background: {color_borde};
                    height: 20px;
                    border-radius: 10px;
                ">
                </div>

            </div>

            <div style="
                font-size: 13px;
                color: #52514e;
                margin-top: 6px;
            ">

                {porcentaje_usado:.0f}% utilizado

            </div>

        </div>
        """))


        # -------------------------------------------------
        # 16. MOSTRAR HISTORIAL DE ESTA SIMULACIÓN
        # -------------------------------------------------

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            margin-top: 25px;
            margin-bottom: 8px;
        ">

            <div style="
                font-size: 22px;
                font-weight: bold;
                color: #1F4E79;
            ">
                🧾 Historial de gastos
            </div>

            <div style="
                font-size: 13px;
                color: #52514e;
                margin-top: 4px;
            ">

                👤 <b>{estudiante}</b>

                &nbsp; | &nbsp;

                📅 <b>{mes}</b>

                <br><br>

                Aquí aparecen únicamente los gastos
                de esta simulación.

            </div>

        </div>
        """))


        display(

            gastos_df.style

            .format({

                "Valor":
                "${:,.0f}"

            })

            .set_caption(
                "Gastos registrados hasta el momento"
            )

        )


        # -------------------------------------------------
        # 17. CONTAR LOS GASTOS REGISTRADOS
        # -------------------------------------------------

        numero_gastos = len(gastos)


        # -------------------------------------------------
        # 18. MOSTRAR INSTRUCCIONES PARA CONTINUAR
        # -------------------------------------------------

        display(HTML(f"""
        <div style="
            font-family: Arial, sans-serif;
            background: #eef4fc;
            border-left: 6px solid #2a78d6;
            border-radius: 8px;
            padding: 16px;
            margin-top: 18px;
            color: #1F4E79;
        ">

            <div style="
                font-size: 19px;
                font-weight: bold;
                margin-bottom: 7px;
            ">
                ➕ ¿Quieres registrar otro gasto?
            </div>

            <div style="
                font-size: 14px;
                color: #303030;
                line-height: 1.8;
            ">

                Actualmente tienes

                <b>{numero_gastos} gasto(s)</b>

                registrado(s) en esta simulación.

                <br><br>

                Para registrar otro:

                <br>

                <b>1.</b>
                Mantén seleccionada
                <b>Registrar gasto</b>.

                <br>

                <b>2.</b>
                Mantén el mismo estudiante y el mismo mes.

                <br>

                <b>3.</b>
                Cambia Categoría, Descripción y Valor.

                <br>

                <b>4.</b>
                Vuelve a presionar ▶️.

                <br><br>

                El gasto nuevo se agregará al historial
                <b>sin borrar los anteriores.</b>

            </div>

        </div>
        """))
```

```python
#@title 5️⃣ CIERRE DEL MES: PLAN VS. REALIDAD { display-mode: "form" }

#@markdown ### 🏁 Cierra tu mes financiero
#@markdown Si registraste todos tus gastos en la Celda 3, selecciona **Usar gastos registrados**.

modo_cierre = "Usar gastos registrados" #@param ["Usar gastos registrados", "Ingresar valores finales manualmente"]

#@markdown ### ✏️ Gastos finales manuales
#@markdown Estos campos SOLO se usan si seleccionas **Ingresar valores finales manualmente**.

transporte_real_manual = 0 #@param {type:"number"}
alimentacion_real_manual = 0 #@param {type:"number"}
materiales_real_manual = 0 #@param {type:"number"}
deseos_real_manual = 0 #@param {type:"number"}


# ---------------------------------------------------------
# 1. VERIFICAR QUE EXISTA EL PLAN FINANCIERO
# ---------------------------------------------------------

cierre_valido = True


if "mensualidad" not in globals():

    cierre_valido = False

    display(HTML("""
    <div style="
        font-family: Arial, sans-serif;
        background:#fff3f3;
        border-left:6px solid #d9534f;
        border-radius:8px;
        padding:16px;
        color:#8a1f1f;
    ">

        <div style="
            font-size:20px;
            font-weight:bold;
            margin-bottom:7px;
        ">
            ⚠️ Primero debes crear tu plan financiero
        </div>

        Ejecuta primero la <b>Celda 2</b>.

    </div>
    """))


# ---------------------------------------------------------
# 2. VERIFICAR QUE EXISTA ESTUDIANTE Y MES
# ---------------------------------------------------------

if cierre_valido:

    if (
        "estudiante" not in globals()
        or "mes" not in globals()
    ):

        cierre_valido = False

        display(HTML("""
        <div style="
            font-family: Arial, sans-serif;
            background:#fff3f3;
            border-left:6px solid #d9534f;
            border-radius:8px;
            padding:16px;
            color:#8a1f1f;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                ⚠️ No existe un seguimiento activo
            </div>

            Ejecuta primero la <b>Celda 3</b>
            para identificar al estudiante y el mes.

        </div>
        """))


# ---------------------------------------------------------
# 3. VALIDAR LOS VALORES MANUALES
# ---------------------------------------------------------

if (
    cierre_valido
    and modo_cierre == "Ingresar valores finales manualmente"
):

    valores_manuales = [
        transporte_real_manual,
        alimentacion_real_manual,
        materiales_real_manual,
        deseos_real_manual
    ]


    if min(valores_manuales) < 0:

        cierre_valido = False

        display(HTML("""
        <div style="
            font-family: Arial, sans-serif;
            background:#fff3f3;
            border-left:6px solid #d9534f;
            border-radius:8px;
            padding:16px;
            color:#8a1f1f;
        ">

            ⚠️ Los gastos no pueden ser negativos.

        </div>
        """))


# ---------------------------------------------------------
# 4. OBTENER LOS GASTOS REALES
# ---------------------------------------------------------

if cierre_valido:


    # -----------------------------------------------------
    # OPCIÓN A: USAR GASTOS REGISTRADOS
    # -----------------------------------------------------

    if modo_cierre == "Usar gastos registrados":


        if "gastos" not in globals() or len(gastos) == 0:

            cierre_valido = False

            display(HTML(f"""
            <div style="
                font-family:Arial,sans-serif;
                background:#fffbea;
                border-left:6px solid #e0a800;
                border-radius:8px;
                padding:16px;
                color:#765700;
            ">

                <div style="
                    font-size:20px;
                    font-weight:bold;
                    margin-bottom:7px;
                ">
                    📭 No hay gastos registrados
                </div>

                👤 <b>{estudiante}</b>
                &nbsp; | &nbsp;
                📅 <b>{mes}</b>

                <br><br>

                Registra gastos en la <b>Celda 3</b>
                o selecciona
                <b>Ingresar valores finales manualmente</b>.

            </div>
            """))


        else:

            gastos_cierre_df = pd.DataFrame(
                gastos,
                columns=[
                    "Categoría",
                    "Descripción",
                    "Valor"
                ]
            )


            totales_reales = (
                gastos_cierre_df
                .groupby("Categoría")["Valor"]
                .sum()
            )


            transporte_real = totales_reales.get(
                "Transporte",
                0
            )


            alimentacion_real = totales_reales.get(
                "Alimentación",
                0
            )


            materiales_real = totales_reales.get(
                "Materiales",
                0
            )


            deseos_real = totales_reales.get(
                "Deseos",
                0
            )


    # -----------------------------------------------------
    # OPCIÓN B: INGRESAR GASTOS MANUALMENTE
    # -----------------------------------------------------

    else:

        transporte_real = transporte_real_manual
        alimentacion_real = alimentacion_real_manual
        materiales_real = materiales_real_manual
        deseos_real = deseos_real_manual


# ---------------------------------------------------------
# 5. CONTINUAR SOLO SI TODO ES CORRECTO
# ---------------------------------------------------------

if cierre_valido:


    # -----------------------------------------------------
    # 6. CALCULAR GASTOS REALES TOTALES
    # -----------------------------------------------------

    total_gastos_real = (
        transporte_real
        + alimentacion_real
        + materiales_real
        + deseos_real
    )


    # -----------------------------------------------------
    # 7. CALCULAR AHORRO REAL AUTOMÁTICAMENTE
    # -----------------------------------------------------

    saldo_despues_de_gastos = (
        mensualidad
        - total_gastos_real
    )


    if saldo_despues_de_gastos >= 0:

        ahorro_real = saldo_despues_de_gastos

        deficit = 0


    else:

        ahorro_real = 0

        deficit = abs(
            saldo_despues_de_gastos
        )


    # -----------------------------------------------------
    # 8. CALCULAR PORCENTAJE REAL DE AHORRO
    # -----------------------------------------------------

    if mensualidad > 0:

        porcentaje_ahorro_real = (
            ahorro_real
            / mensualidad
        ) * 100

    else:

        porcentaje_ahorro_real = 0


    # -----------------------------------------------------
    # 9. DEFINIR LAS CATEGORÍAS
    # -----------------------------------------------------

    categorias_finales = [
        "Transporte",
        "Alimentación",
        "Materiales",
        "Deseos",
        "Ahorro"
    ]


    # -----------------------------------------------------
    # 10. ORGANIZAR PLAN Y REALIDAD CON NUMPY
    # -----------------------------------------------------

    plan_array = np.array([
        transporte,
        alimentacion,
        materiales,
        deseos,
        ahorro
    ])


    real_array = np.array([
        transporte_real,
        alimentacion_real,
        materiales_real,
        deseos_real,
        ahorro_real
    ])


    # -----------------------------------------------------
    # 11. CALCULAR DIFERENCIAS
    # -----------------------------------------------------

    diferencias = (
        real_array
        - plan_array
    )


    # -----------------------------------------------------
    # 12. FUNCIÓN PARA EVALUAR CADA RESULTADO
    # -----------------------------------------------------

    def evaluar_resultado(categoria, plan, real):


        # AHORRO

        if categoria == "Ahorro":

            if plan > 0:

                porcentaje = real / plan

            else:

                porcentaje = 0


            if real >= plan:

                estado = "✅ Meta alcanzada"

            elif real >= plan * 0.85:

                estado = "⚠️ Cerca de la meta"

            else:

                estado = "🔔 Meta no alcanzada"


        # GASTOS

        else:

            if plan > 0:

                porcentaje = real / plan

            else:

                if real > 0:
                    porcentaje = 1
                else:
                    porcentaje = 0


            if real <= plan:

                estado = "✅ En presupuesto"

            elif real <= plan * 1.10:

                estado = "⚠️ Exceso leve"

            else:

                estado = "🚨 Presupuesto excedido"


        return estado, porcentaje


    # -----------------------------------------------------
    # 13. EVALUAR TODAS LAS CATEGORÍAS
    # -----------------------------------------------------

    estados_finales = []

    porcentajes_finales = []


    for i in range(
        len(categorias_finales)
    ):

        estado, porcentaje = evaluar_resultado(

            categorias_finales[i],

            plan_array[i],

            real_array[i]

        )


        estados_finales.append(
            estado
        )


        porcentajes_finales.append(
            porcentaje
        )


    # -----------------------------------------------------
    # 14. CREAR TABLA PLAN VS. REALIDAD
    # -----------------------------------------------------

    resultado_final = pd.DataFrame({

        "Categoría":
        categorias_finales,

        "Plan":
        plan_array,

        "Real":
        real_array,

        "Diferencia":
        diferencias,

        "% del plan":
        porcentajes_finales,

        "Resultado":
        estados_finales

    })


    # -----------------------------------------------------
    # 15. CALCULAR TOTAL DE GASTOS PLANEADOS
    # -----------------------------------------------------

    total_gastos_plan = (
        transporte
        + alimentacion
        + materiales
        + deseos
    )


    # -----------------------------------------------------
    # 16. CALCULAR CUÁNTAS METAS SE CUMPLIERON
    # -----------------------------------------------------

    metas_cumplidas = 0


    for estado in estados_finales:

        if estado.startswith("✅"):

            metas_cumplidas = (
                metas_cumplidas
                + 1
            )


    # -----------------------------------------------------
    # 17. MOSTRAR ENCABEZADO
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        margin-top:10px;
        margin-bottom:20px;
    ">

        <div style="
            font-size:29px;
            font-weight:bold;
            color:#1F4E79;
        ">
            🏁 Cierre financiero del mes
        </div>


        <div style="
            font-size:15px;
            color:#303030;
            margin-top:8px;
        ">

            👤 <b>{estudiante}</b>

            &nbsp;&nbsp; | &nbsp;&nbsp;

            📅 <b>{mes}</b>

        </div>


        <div style="
            font-size:14px;
            color:#52514e;
            margin-top:7px;
            line-height:1.7;
        ">

            Así terminó tu mes comparado con el plan
            financiero que creaste al inicio.

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 18. MOSTRAR TARJETAS PRINCIPALES
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        display:flex;
        gap:12px;
        flex-wrap:wrap;
        margin-bottom:25px;
    ">


        <div style="
            flex:1;
            min-width:175px;
            background:#f5f7fa;
            border-top:5px solid #2a78d6;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💰 Mensualidad
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1F4E79;
                margin-top:4px;
            ">
                ${mensualidad:,.0f}
            </div>

            <div style="
                font-size:11px;
                color:#52514e;
            ">
                Dinero disponible al iniciar el mes
            </div>

        </div>


        <div style="
            flex:1;
            min-width:175px;
            background:#f5f7fa;
            border-top:5px solid #eb6834;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💸 Gastos reales
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#eb6834;
                margin-top:4px;
            ">
                ${total_gastos_real:,.0f}
            </div>

            <div style="
                font-size:11px;
                color:#52514e;
            ">
                Plan para gastos:
                ${total_gastos_plan:,.0f}
            </div>

        </div>


        <div style="
            flex:1;
            min-width:175px;
            background:#f5f7fa;
            border-top:5px solid #1baf7a;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💚 Ahorro real
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1baf7a;
                margin-top:4px;
            ">
                ${ahorro_real:,.0f}
            </div>

            <div style="
                font-size:11px;
                color:#52514e;
            ">
                {porcentaje_ahorro_real:.0f}% de tu mensualidad
            </div>

        </div>


        <div style="
            flex:1;
            min-width:175px;
            background:#f5f7fa;
            border-top:5px solid #7b61a8;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                🎯 Meta de ahorro
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#7b61a8;
                margin-top:4px;
            ">
                ${ahorro:,.0f}
            </div>

            <div style="
                font-size:11px;
                color:#52514e;
            ">
                20% planeado al comenzar el mes
            </div>

        </div>


        <div style="
            flex:1;
            min-width:175px;
            background:#f5f7fa;
            border-top:5px solid #52514e;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                ⭐ Metas cumplidas
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#303030;
                margin-top:4px;
            ">
                {metas_cumplidas}/5
            </div>

            <div style="
                font-size:11px;
                color:#52514e;
            ">
                Objetivos financieros cumplidos
            </div>

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 19. MOSTRAR RESULTADO DEL AHORRO
    # -----------------------------------------------------

    if deficit > 0:

        display(HTML(f"""
        <div style="
            font-family:Arial,sans-serif;
            background:#fff3f3;
            border-left:6px solid #d9534f;
            border-radius:8px;
            padding:17px;
            margin-bottom:20px;
            color:#8a1f1f;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                🚨 Terminaste el mes con déficit
            </div>

            Tus gastos superaron tu mensualidad en

            <b>${deficit:,.0f}</b>.

            <br><br>

            Tu ahorro real fue de
            <b>$0</b>.

        </div>
        """))


    elif ahorro_real >= ahorro:

        ahorro_extra = (
            ahorro_real
            - ahorro
        )


        display(HTML(f"""
        <div style="
            font-family:Arial,sans-serif;
            background:#eef8f4;
            border-left:6px solid #1baf7a;
            border-radius:8px;
            padding:17px;
            margin-bottom:20px;
            color:#184f3d;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                🎉 ¡Cumpliste tu meta de ahorro!
            </div>

            Planeabas ahorrar:

            <b>${ahorro:,.0f}</b>

            <br>

            Ahorraste realmente:

            <b>${ahorro_real:,.0f}</b>

            <br><br>

            Superaste tu meta en:

            <b>${ahorro_extra:,.0f}</b>

        </div>
        """))


    else:

        faltante_ahorro = (
            ahorro
            - ahorro_real
        )


        display(HTML(f"""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:17px;
            margin-bottom:20px;
            color:#765700;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                🎯 Estuviste cerca de tu meta de ahorro
            </div>

            Planeabas ahorrar:

            <b>${ahorro:,.0f}</b>

            <br>

            Ahorraste realmente:

            <b>${ahorro_real:,.0f}</b>

            <br><br>

            Te faltaron:

            <b>${faltante_ahorro:,.0f}</b>

            para alcanzar tu meta.

        </div>
        """))


    # -----------------------------------------------------
    # 20. MOSTRAR TABLA PLAN VS. REALIDAD
    # -----------------------------------------------------

    display(HTML("""
    <div style="
        font-family:Arial,sans-serif;
        font-size:22px;
        font-weight:bold;
        color:#1F4E79;
        margin-bottom:8px;
    ">
        📋 Plan vs. realidad
    </div>
    """))


    display(

        resultado_final.style

        .format({

            "Plan":
            "${:,.0f}",

            "Real":
            "${:,.0f}",

            "Diferencia":
            "${:+,.0f}",

            "% del plan":
            "{:.0%}"

        })

        .set_caption(

            f"Resultado financiero de {estudiante} - {mes}"

        )

    )


    # -----------------------------------------------------
    # 21. CALCULAR MARGEN FAVORABLE
    # -----------------------------------------------------

    margen_favorable = np.array([

        transporte - transporte_real,

        alimentacion - alimentacion_real,

        materiales - materiales_real,

        deseos - deseos_real,

        ahorro_real - ahorro

    ])


    # -----------------------------------------------------
    # 22. GRÁFICA 1:
    # PLAN VS. REALIDAD
    # -----------------------------------------------------

    posiciones = np.arange(
        len(categorias_finales)
    )


    ancho = 0.38


    plt.figure(
        figsize=(11, 5)
    )


    plt.bar(

        posiciones - ancho / 2,

        plan_array,

        width=ancho,

        color="#2a78d6",

        label="Plan"

    )


    plt.bar(

        posiciones + ancho / 2,

        real_array,

        width=ancho,

        color="#eb6834",

        label="Real"

    )


    plt.xticks(
        posiciones,
        categorias_finales
    )


    plt.title(
        "Plan vs. realidad del mes",
        loc="left",
        fontsize=16,
        fontweight="bold"
    )


    plt.ylabel(
        "Pesos colombianos"
    )


    plt.legend(
        frameon=False
    )


    plt.gca().spines[
        ["top", "right"]
    ].set_visible(False)


    plt.tight_layout()

    plt.show()


    # -----------------------------------------------------
    # 23. GRÁFICA 2:
    # RESULTADO FRENTE AL PLAN
    # -----------------------------------------------------

    colores_resultado = []


    for margen in margen_favorable:

        if margen >= 0:

            colores_resultado.append(
                "#1baf7a"
            )

        else:

            colores_resultado.append(
                "#d9534f"
            )


    plt.figure(
        figsize=(11, 5)
    )


    barras_margen = plt.bar(

        categorias_finales,

        margen_favorable,

        color=colores_resultado

    )


    plt.axhline(
        0,
        color="#52514e",
        linewidth=1
    )


    plt.title(
        "Resultado frente al plan",
        loc="left",
        fontsize=16,
        fontweight="bold"
    )


    plt.ylabel(
        "Margen favorable / desfavorable"
    )


    plt.gca().spines[
        ["top", "right"]
    ].set_visible(False)


    plt.bar_label(

        barras_margen,

        labels=[
            f"${valor:+,.0f}"
            for valor in margen_favorable
        ],

        padding=3

    )


    plt.tight_layout()

    plt.show()


    # -----------------------------------------------------
    # 24. CREAR RECOMENDACIONES PERSONALIZADAS
    # -----------------------------------------------------

    recomendaciones = []


    # TRANSPORTE

    if transporte_real > transporte:

        exceso = (
            transporte_real
            - transporte
        )

        recomendaciones.append(
            f"""
            🚍 <b>Transporte:</b>
            gastaste <b>${exceso:,.0f}</b>
            más de lo planeado.
            Puedes revisar tus trayectos frecuentes
            y establecer un límite semanal.
            """
        )

    else:

        diferencia = (
            transporte
            - transporte_real
        )

        recomendaciones.append(
            f"""
            🚍 <b>Transporte:</b>
            gastaste <b>${diferencia:,.0f}</b>
            menos de lo presupuestado.
            """
        )


    # ALIMENTACIÓN

    if alimentacion_real > alimentacion:

        exceso = (
            alimentacion_real
            - alimentacion
        )

        recomendaciones.append(
            f"""
            🍴 <b>Alimentación:</b>
            superaste tu presupuesto en
            <b>${exceso:,.0f}</b>.
            Un límite semanal puede ayudarte
            a controlar mejor los gastos frecuentes.
            """
        )

    else:

        diferencia = (
            alimentacion
            - alimentacion_real
        )

        recomendaciones.append(
            f"""
            🍴 <b>Alimentación:</b>
            terminaste
            <b>${diferencia:,.0f}</b>
            por debajo de tu presupuesto.
            """
        )


    # MATERIALES

    if materiales_real > materiales:

        exceso = (
            materiales_real
            - materiales
        )

        recomendaciones.append(
            f"""
            📚 <b>Materiales:</b>
            superaste el presupuesto en
            <b>${exceso:,.0f}</b>.
            Puedes anticipar impresiones,
            fotocopias y materiales académicos
            para el próximo mes.
            """
        )

    else:

        diferencia = (
            materiales
            - materiales_real
        )

        recomendaciones.append(
            f"""
            📚 <b>Materiales:</b>
            gastaste
            <b>${diferencia:,.0f}</b>
            menos de lo planeado.
            """
        )


    # DESEOS

    if deseos_real > deseos:

        exceso = (
            deseos_real
            - deseos
        )

        recomendaciones.append(
            f"""
            ✨ <b>Deseos:</b>
            superaste tu presupuesto en
            <b>${exceso:,.0f}</b>.
            Puedes definir un límite semanal
            para ocio y compras no esenciales.
            """
        )

    else:

        diferencia = (
            deseos
            - deseos_real
        )

        recomendaciones.append(
            f"""
            ✨ <b>Deseos:</b>
            terminaste
            <b>${diferencia:,.0f}</b>
            por debajo del presupuesto.
            """
        )


    # AHORRO

    if ahorro_real >= ahorro:

        ahorro_extra = (
            ahorro_real
            - ahorro
        )

        recomendaciones.append(
            f"""
            💚 <b>Ahorro:</b>
            cumpliste tu meta y lograste ahorrar
            <b>${ahorro_real:,.0f}</b>.

            Superaste tu objetivo en
            <b>${ahorro_extra:,.0f}</b>.
            """
        )

    else:

        faltante = (
            ahorro
            - ahorro_real
        )

        recomendaciones.append(
            f"""
            💚 <b>Ahorro:</b>
            ahorraste
            <b>${ahorro_real:,.0f}</b>.

            Te faltaron
            <b>${faltante:,.0f}</b>
            para alcanzar la meta del mes.
            """
        )


    # DÉFICIT

    if deficit > 0:

        recomendaciones.append(
            f"""
            🚨 Terminaste el mes gastando
            <b>${deficit:,.0f}</b>
            más de tu mensualidad.
            El próximo mes debes priorizar
            las categorías que más se excedieron.
            """
        )


    # -----------------------------------------------------
    # 25. CONVERTIR RECOMENDACIONES A HTML
    # -----------------------------------------------------

    recomendaciones_html = ""


    for recomendacion in recomendaciones:

        recomendaciones_html = (
            recomendaciones_html
            + f"""
            <div style="
                margin-bottom:14px;
                line-height:1.7;
            ">
                {recomendacion}
            </div>
            """
        )


    # -----------------------------------------------------
    # 26. MOSTRAR RECOMENDACIONES
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        background:#eef4fc;
        border-left:6px solid #2a78d6;
        border-radius:8px;
        padding:18px;
        margin-top:20px;
    ">

        <div style="
            font-size:22px;
            font-weight:bold;
            color:#1F4E79;
            margin-bottom:12px;
        ">
            💡 Recomendaciones para tu próximo mes
        </div>

        <div style="
            font-size:14px;
            color:#303030;
        ">

            {recomendaciones_html}

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 27. GENERAR MENSAJE FINAL
    # -----------------------------------------------------

    if (
        metas_cumplidas == 5
        and deficit == 0
    ):

        mensaje_cierre = """
        🎉 ¡Excelente trabajo!

        Cumpliste todas las metas de tu plan financiero.
        Sigue usando esta información para mantener
        buenos hábitos financieros.
        """

        fondo_cierre = "#eef8f4"
        borde_cierre = "#1baf7a"
        texto_cierre = "#184f3d"


    elif (
        metas_cumplidas >= 3
        and deficit == 0
    ):

        mensaje_cierre = """
        👏 Tuviste un buen mes.

        Cumpliste la mayoría de tus objetivos.
        Ahora puedes ajustar las categorías que tuvieron
        mayores diferencias frente al plan.
        """

        fondo_cierre = "#fffbea"
        borde_cierre = "#e0a800"
        texto_cierre = "#765700"


    else:

        mensaje_cierre = """
        📚 Este mes te dejó información muy útil.

        Ahora sabes exactamente en qué categorías debes
        ajustar tus gastos para construir un plan financiero
        más adecuado el próximo mes.
        """

        fondo_cierre = "#fff4e8"
        borde_cierre = "#eb6834"
        texto_cierre = "#9a431f"


    # -----------------------------------------------------
    # 28. MOSTRAR RESULTADO FINAL
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        background:{fondo_cierre};
        border-left:6px solid {borde_cierre};
        border-radius:8px;
        padding:19px;
        margin-top:20px;
        color:{texto_cierre};
    ">

        <div style="
            font-size:22px;
            font-weight:bold;
            margin-bottom:8px;
        ">
            🌟 Resultado del mes
        </div>

        <div style="
            font-size:15px;
            line-height:1.8;
        ">

            {mensaje_cierre}

        </div>

    </div>
    """))
```

```python
#@title 6️⃣ GUARDA TU MES FINANCIERO { display-mode: "form" }

#@markdown ### 💾 Guarda toda la información de tu mes
#@markdown Se creará un archivo Excel con:
#@markdown - **Historial de gastos**
#@markdown - **Resumen financiero final**


# ---------------------------------------------------------
# 1. IMPORTAR HERRAMIENTAS NECESARIAS
# ---------------------------------------------------------

from google.colab import files as colab_files

from IPython.display import (
    display,
    HTML,
    FileLink
)

from openpyxl import load_workbook

from openpyxl.styles import (
    Font,
    PatternFill,
    Alignment
)


# ---------------------------------------------------------
# 2. COMENZAR VALIDACIÓN
# ---------------------------------------------------------

exportacion_valida = True


# ---------------------------------------------------------
# 3. VERIFICAR QUE EXISTA EL PLAN FINANCIERO
# ---------------------------------------------------------

variables_plan = [
    "mensualidad",
    "transporte",
    "alimentacion",
    "materiales",
    "deseos",
    "ahorro"
]


for variable in variables_plan:

    if variable not in globals():

        exportacion_valida = False


if not exportacion_valida:

    display(HTML("""
    <div style="
        font-family:Arial,sans-serif;
        background:#fff3f3;
        border-left:6px solid #d9534f;
        border-radius:8px;
        padding:16px;
        margin-top:10px;
        color:#8a1f1f;
    ">

        <div style="
            font-size:20px;
            font-weight:bold;
            margin-bottom:7px;
        ">
            ⚠️ No existe un plan financiero completo
        </div>

        Ejecuta primero la <b>Celda 2</b>
        para crear tu presupuesto.

    </div>
    """))


# ---------------------------------------------------------
# 4. VERIFICAR QUE EXISTA ESTUDIANTE Y MES
# ---------------------------------------------------------

if exportacion_valida:

    if (
        "estudiante" not in globals()
        or "mes" not in globals()
    ):

        exportacion_valida = False

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fff3f3;
            border-left:6px solid #d9534f;
            border-radius:8px;
            padding:16px;
            margin-top:10px;
            color:#8a1f1f;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                ⚠️ No existe un seguimiento activo
            </div>

            Ejecuta primero la <b>Celda 3</b>
            para identificar al estudiante y el mes.

        </div>
        """))


# ---------------------------------------------------------
# 5. VERIFICAR QUE SE HAYA CERRADO EL MES EN CELDA 5
# ---------------------------------------------------------

if exportacion_valida:

    if (
        "modo_cierre" not in globals()
        or "resultado_final" not in globals()
    ):

        exportacion_valida = False

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:16px;
            margin-top:10px;
            color:#765700;
        ">

            <div style="
                font-size:20px;
                font-weight:bold;
                margin-bottom:7px;
            ">
                🏁 Primero debes cerrar el mes
            </div>

            Ejecuta la <b>Celda 5</b>
            antes de guardar tu información.

            <br><br>

            Así nos aseguramos de guardar exactamente
            el cierre financiero que acabas de revisar.

        </div>
        """))


# ---------------------------------------------------------
# 6. OBTENER LOS VALORES REALES SEGÚN EL MODO DE CIERRE
# ---------------------------------------------------------

if exportacion_valida:


    # =====================================================
    # OPCIÓN A:
    # USAR LOS GASTOS REGISTRADOS
    # =====================================================

    if modo_cierre == "Usar gastos registrados":


        if (
            "gastos" not in globals()
            or len(gastos) == 0
        ):

            exportacion_valida = False

            display(HTML("""
            <div style="
                font-family:Arial,sans-serif;
                background:#fffbea;
                border-left:6px solid #e0a800;
                border-radius:8px;
                padding:16px;
                color:#765700;
            ">

                ⚠️ Seleccionaste
                <b>Usar gastos registrados</b>,
                pero actualmente no hay gastos registrados.

                <br><br>

                Revisa la <b>Celda 3</b>
                y vuelve a ejecutar la <b>Celda 5</b>.

            </div>
            """))


        else:


            # Crear tabla con los gastos actuales

            gastos_exportacion_df = pd.DataFrame(

                gastos,

                columns=[
                    "Categoría",
                    "Descripción",
                    "Valor"
                ]

            )


            # Sumar gastos por categoría

            totales_exportacion = (

                gastos_exportacion_df

                .groupby("Categoría")["Valor"]

                .sum()

            )


            transporte_exportar = totales_exportacion.get(
                "Transporte",
                0
            )


            alimentacion_exportar = totales_exportacion.get(
                "Alimentación",
                0
            )


            materiales_exportar = totales_exportacion.get(
                "Materiales",
                0
            )


            deseos_exportar = totales_exportacion.get(
                "Deseos",
                0
            )


    # =====================================================
    # OPCIÓN B:
    # USAR LOS VALORES FINALES MANUALES
    # =====================================================

    else:


        variables_manuales = [

            "transporte_real_manual",

            "alimentacion_real_manual",

            "materiales_real_manual",

            "deseos_real_manual"

        ]


        manuales_disponibles = True


        for variable in variables_manuales:

            if variable not in globals():

                manuales_disponibles = False


        if not manuales_disponibles:

            exportacion_valida = False

            display(HTML("""
            <div style="
                font-family:Arial,sans-serif;
                background:#fffbea;
                border-left:6px solid #e0a800;
                border-radius:8px;
                padding:16px;
                color:#765700;
            ">

                ⚠️ No encontramos los valores manuales
                utilizados para cerrar el mes.

                <br><br>

                Ejecuta nuevamente la <b>Celda 5</b>
                antes de guardar.

            </div>
            """))


        else:

            transporte_exportar = (
                transporte_real_manual
            )

            alimentacion_exportar = (
                alimentacion_real_manual
            )

            materiales_exportar = (
                materiales_real_manual
            )

            deseos_exportar = (
                deseos_real_manual
            )


# ---------------------------------------------------------
# 7. CALCULAR RESULTADOS FINALES
# ---------------------------------------------------------

if exportacion_valida:


    total_gastos_exportar = (

        transporte_exportar

        + alimentacion_exportar

        + materiales_exportar

        + deseos_exportar

    )


    # -----------------------------------------------------
    # CALCULAR AHORRO REAL
    # -----------------------------------------------------

    saldo_exportar = (

        mensualidad

        - total_gastos_exportar

    )


    if saldo_exportar >= 0:

        ahorro_exportar = (
            saldo_exportar
        )

        deficit_exportar = 0


    else:

        ahorro_exportar = 0

        deficit_exportar = abs(
            saldo_exportar
        )


    # -----------------------------------------------------
    # PORCENTAJE REAL DE AHORRO
    # -----------------------------------------------------

    if mensualidad > 0:

        porcentaje_ahorro_exportar = (

            ahorro_exportar
            / mensualidad

        ) * 100


    else:

        porcentaje_ahorro_exportar = 0


# ---------------------------------------------------------
# 8. CREAR PLAN Y REALIDAD PARA COMPROBAR CELDA 5
# ---------------------------------------------------------

if exportacion_valida:


    categorias_exportar = [

        "Transporte",

        "Alimentación",

        "Materiales",

        "Deseos",

        "Ahorro"

    ]


    plan_exportar = [

        transporte,

        alimentacion,

        materiales,

        deseos,

        ahorro

    ]


    realidad_exportar = [

        transporte_exportar,

        alimentacion_exportar,

        materiales_exportar,

        deseos_exportar,

        ahorro_exportar

    ]


# ---------------------------------------------------------
# 9. COMPROBAR QUE LA CELDA 5 ESTÉ ACTUALIZADA
# ---------------------------------------------------------

if exportacion_valida:


    cierre_coincide = True


    try:


        for i in range(
            len(categorias_exportar)
        ):


            categoria_actual = (
                categorias_exportar[i]
            )


            valor_esperado = (
                realidad_exportar[i]
            )


            fila_resultado = resultado_final[

                resultado_final["Categoría"]
                == categoria_actual

            ]


            valor_celda5 = float(

                fila_resultado[
                    "Real"
                ].iloc[0]

            )


            if abs(
                valor_celda5
                - valor_esperado
            ) > 0.01:

                cierre_coincide = False


    except:

        cierre_coincide = False


    # -----------------------------------------------------
    # SI NO COINCIDE, NO GUARDAR
    # -----------------------------------------------------

    if not cierre_coincide:

        exportacion_valida = False

        display(HTML(f"""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:18px;
            margin-top:12px;
            color:#765700;
        ">

            <div style="
                font-size:21px;
                font-weight:bold;
                margin-bottom:8px;
            ">
                🔄 Tu cierre necesita actualizarse
            </div>

            <div style="
                font-size:14px;
                line-height:1.8;
            ">

                Detectamos que los datos actuales de

                <b>{estudiante}</b>

                para

                <b>{mes}</b>

                no coinciden con el último cierre
                realizado en la Celda 5.

                <br><br>

                Esto puede ocurrir si registraste o cambiaste
                gastos después de cerrar el mes.

                <br><br>

                👉 Ejecuta nuevamente la
                <b>Celda 5</b>

                y después vuelve a ejecutar esta
                <b>Celda 6</b>.

                <br><br>

                Así evitamos guardar información incorrecta.

            </div>

        </div>
        """))


# ---------------------------------------------------------
# 10. SI TODO COINCIDE, CREAR EL EXCEL
# ---------------------------------------------------------

if exportacion_valida:


    # =====================================================
    # CREAR HISTORIAL DE GASTOS
    # =====================================================

    if (
        "gastos" in globals()
        and len(gastos) > 0
    ):


        historial_exportar = pd.DataFrame(

            gastos,

            columns=[
                "Categoría",
                "Descripción",
                "Valor"
            ]

        )


    else:


        historial_exportar = pd.DataFrame(

            columns=[
                "Categoría",
                "Descripción",
                "Valor"
            ]

        )


    # Agregar estudiante y mes

    historial_exportar.insert(

        0,

        "Estudiante",

        estudiante

    )


    historial_exportar.insert(

        1,

        "Mes",

        mes

    )


    # -----------------------------------------------------
    # 11. CALCULAR DIFERENCIAS
    # -----------------------------------------------------

    diferencias_exportar = []


    for i in range(
        len(categorias_exportar)
    ):


        diferencia = (

            realidad_exportar[i]

            - plan_exportar[i]

        )


        diferencias_exportar.append(
            diferencia
        )


    # -----------------------------------------------------
    # 12. CALCULAR PORCENTAJE Y RESULTADO
    # -----------------------------------------------------

    porcentajes_exportar = []

    resultados_exportar = []


    for i in range(
        len(categorias_exportar)
    ):


        categoria_actual = (
            categorias_exportar[i]
        )


        plan_actual = (
            plan_exportar[i]
        )


        real_actual = (
            realidad_exportar[i]
        )


        # -------------------------------------------------
        # PORCENTAJE
        # -------------------------------------------------

        if plan_actual > 0:

            porcentaje = (

                real_actual
                / plan_actual

            )


        else:

            if real_actual > 0:

                porcentaje = 1

            else:

                porcentaje = 0


        porcentajes_exportar.append(
            porcentaje
        )


        # -------------------------------------------------
        # AHORRO
        # -------------------------------------------------

        if categoria_actual == "Ahorro":


            if real_actual >= plan_actual:

                resultado = (
                    "Meta alcanzada"
                )


            elif real_actual >= plan_actual * 0.85:

                resultado = (
                    "Cerca de la meta"
                )


            else:

                resultado = (
                    "Meta no alcanzada"
                )


        # -------------------------------------------------
        # GASTOS
        # -------------------------------------------------

        else:


            if real_actual <= plan_actual:

                resultado = (
                    "En presupuesto"
                )


            elif real_actual <= plan_actual * 1.10:

                resultado = (
                    "Exceso leve"
                )


            else:

                resultado = (
                    "Presupuesto excedido"
                )


        resultados_exportar.append(
            resultado
        )


    # -----------------------------------------------------
    # 13. CREAR RESUMEN FINANCIERO
    # -----------------------------------------------------

    resumen_exportar = pd.DataFrame({

        "Categoría":
        categorias_exportar,

        "Plan":
        plan_exportar,

        "Real":
        realidad_exportar,

        "Diferencia":
        diferencias_exportar,

        "% del plan":
        porcentajes_exportar,

        "Resultado":
        resultados_exportar

    })


    # -----------------------------------------------------
    # 14. CREAR NOMBRE DEL ARCHIVO
    # -----------------------------------------------------

    nombre_estudiante = (

        estudiante
        .strip()
        .replace(" ", "_")

    )


    nombre_mes = (

        mes
        .strip()
        .replace(" ", "_")

    )


    nombre_excel = (

        "Finanzas_"

        + nombre_estudiante

        + "_"

        + nombre_mes

        + ".xlsx"

    )


    # -----------------------------------------------------
    # 15. CREAR ARCHIVO EXCEL
    # -----------------------------------------------------

    with pd.ExcelWriter(

        nombre_excel,

        engine="openpyxl"

    ) as writer:


        # HOJA 1

        historial_exportar.to_excel(

            writer,

            sheet_name="Historial de gastos",

            index=False,

            startrow=4

        )


        # HOJA 2

        resumen_exportar.to_excel(

            writer,

            sheet_name="Resumen financiero",

            index=False,

            startrow=10

        )


    # -----------------------------------------------------
    # 16. ABRIR EXCEL PARA DARLE FORMATO
    # -----------------------------------------------------

    libro = load_workbook(
        nombre_excel
    )


    # Colores

    azul = "2A78D6"

    azul_oscuro = "1F4E79"

    naranja = "EB6834"

    verde = "1BAF7A"

    rojo = "D9534F"

    amarillo = "E0A800"

    gris_claro = "F5F7FA"

    gris = "52514E"

    blanco = "FFFFFF"


    # =====================================================
    # HOJA 1 — HISTORIAL DE GASTOS
    # =====================================================

    hoja_gastos = libro[
        "Historial de gastos"
    ]


    # -----------------------------------------------------
    # TÍTULO
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A1:E1"
    )


    hoja_gastos["A1"] = (
        "Historial de gastos"
    )


    hoja_gastos["A1"].font = Font(

        size=18,

        bold=True,

        color=blanco

    )


    hoja_gastos["A1"].fill = PatternFill(

        "solid",

        fgColor=azul_oscuro

    )


    hoja_gastos["A1"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # ESTUDIANTE Y MES
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A2:E2"
    )


    hoja_gastos["A2"] = (

        estudiante

        + " | "

        + mes

    )


    hoja_gastos["A2"].font = Font(

        italic=True,

        color=gris

    )


    hoja_gastos["A2"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # NOTA SOBRE EL MODO DE CIERRE
    # -----------------------------------------------------

    hoja_gastos.merge_cells(
        "A3:E3"
    )


    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):

        hoja_gastos["A3"] = (

            "Nota: el resumen final usa los valores "
            "manuales ingresados en el cierre. "
            "Este historial muestra únicamente los "
            "gastos registrados en la aplicación."

        )


    else:

        hoja_gastos["A3"] = (

            "El resumen financiero fue calculado "
            "con los gastos registrados en este historial."

        )


    hoja_gastos["A3"].font = Font(

        italic=True,

        size=10,

        color=gris

    )


    hoja_gastos["A3"].alignment = Alignment(

        horizontal="center",

        wrap_text=True

    )


    # -----------------------------------------------------
    # ENCABEZADOS DE TABLA
    # -----------------------------------------------------

    for celda in hoja_gastos[5]:


        celda.font = Font(

            bold=True,

            color=blanco

        )


        celda.fill = PatternFill(

            "solid",

            fgColor=azul

        )


        celda.alignment = Alignment(

            horizontal="center"

        )


    # -----------------------------------------------------
    # FORMATO DE MONEDA
    # -----------------------------------------------------

    for fila in range(

        6,

        hoja_gastos.max_row + 1

    ):


        hoja_gastos[

            f"E{fila}"

        ].number_format = '$#,##0'


    # Congelar encabezados

    hoja_gastos.freeze_panes = "A6"


    # Filtro

    hoja_gastos.auto_filter.ref = (

        f"A5:E{hoja_gastos.max_row}"

    )


    # Ancho de columnas

    hoja_gastos.column_dimensions["A"].width = 18

    hoja_gastos.column_dimensions["B"].width = 15

    hoja_gastos.column_dimensions["C"].width = 20

    hoja_gastos.column_dimensions["D"].width = 42

    hoja_gastos.column_dimensions["E"].width = 18


    # =====================================================
    # HOJA 2 — RESUMEN FINANCIERO
    # =====================================================

    hoja_resumen = libro[
        "Resumen financiero"
    ]


    # -----------------------------------------------------
    # TÍTULO
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A1:F1"
    )


    hoja_resumen["A1"] = (
        "Resumen financiero del mes"
    )


    hoja_resumen["A1"].font = Font(

        size=18,

        bold=True,

        color=blanco

    )


    hoja_resumen["A1"].fill = PatternFill(

        "solid",

        fgColor=azul_oscuro

    )


    hoja_resumen["A1"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # ESTUDIANTE Y MES
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A2:F2"
    )


    hoja_resumen["A2"] = (

        estudiante

        + " | "

        + mes

    )


    hoja_resumen["A2"].font = Font(

        italic=True,

        color=gris

    )


    hoja_resumen["A2"].alignment = Alignment(

        horizontal="center"

    )


    # -----------------------------------------------------
    # INDICADORES GENERALES
    # -----------------------------------------------------

    hoja_resumen["A4"] = (
        "Mensualidad"
    )

    hoja_resumen["B4"] = (
        mensualidad
    )


    hoja_resumen["C4"] = (
        "Total gastado"
    )

    hoja_resumen["D4"] = (
        total_gastos_exportar
    )


    hoja_resumen["A5"] = (
        "Ahorro real"
    )

    hoja_resumen["B5"] = (
        ahorro_exportar
    )


    hoja_resumen["C5"] = (
        "% ahorro"
    )

    hoja_resumen["D5"] = (

        porcentaje_ahorro_exportar
        / 100

    )


    hoja_resumen["A6"] = (
        "Meta de ahorro"
    )

    hoja_resumen["B6"] = (
        ahorro
    )


    hoja_resumen["C6"] = (
        "Déficit"
    )

    hoja_resumen["D6"] = (
        deficit_exportar
    )


    hoja_resumen["A7"] = (
        "Modo de cierre"
    )

    hoja_resumen["B7"] = (
        modo_cierre
    )


    # -----------------------------------------------------
    # NOTA DEL CIERRE
    # -----------------------------------------------------

    hoja_resumen.merge_cells(
        "A8:F8"
    )


    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):

        hoja_resumen["A8"] = (

            "El resumen utiliza los valores finales "
            "ingresados manualmente en la Celda 5."

        )


    else:

        hoja_resumen["A8"] = (

            "El resumen utiliza los gastos registrados "
            "durante el seguimiento."

        )


    hoja_resumen["A8"].font = Font(

        italic=True,

        size=10,

        color=gris

    )


    hoja_resumen["A8"].alignment = Alignment(

        horizontal="center",

        wrap_text=True

    )


    # -----------------------------------------------------
    # FORMATO DE LOS INDICADORES
    # -----------------------------------------------------

    for celda in [

        "A4",
        "C4",
        "A5",
        "C5",
        "A6",
        "C6",
        "A7"

    ]:


        hoja_resumen[
            celda
        ].font = Font(

            bold=True,

            color=azul_oscuro

        )


        hoja_resumen[
            celda
        ].fill = PatternFill(

            "solid",

            fgColor=gris_claro

        )


    # Moneda

    for celda in [

        "B4",
        "D4",
        "B5",
        "B6",
        "D6"

    ]:


        hoja_resumen[
            celda
        ].number_format = '$#,##0'


    # Porcentaje

    hoja_resumen[
        "D5"
    ].number_format = "0%"


    # -----------------------------------------------------
    # ENCABEZADOS PLAN VS. REALIDAD
    # -----------------------------------------------------

    for celda in hoja_resumen[11]:


        celda.font = Font(

            bold=True,

            color=blanco

        )


        celda.fill = PatternFill(

            "solid",

            fgColor=azul

        )


        celda.alignment = Alignment(

            horizontal="center"

        )


    # -----------------------------------------------------
    # FORMATEAR FILAS DEL RESUMEN
    # -----------------------------------------------------

    for fila in range(

        12,

        hoja_resumen.max_row + 1

    ):


        # Moneda

        hoja_resumen[
            f"B{fila}"
        ].number_format = '$#,##0'


        hoja_resumen[
            f"C{fila}"
        ].number_format = '$#,##0'


        hoja_resumen[
            f"D{fila}"
        ].number_format = '$#,##0'


        # Porcentaje

        hoja_resumen[
            f"E{fila}"
        ].number_format = "0%"


        # Resultado

        resultado = hoja_resumen[
            f"F{fila}"
        ].value


        if resultado in [

            "En presupuesto",

            "Meta alcanzada"

        ]:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="E8F5EE"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=verde,

                bold=True

            )


        elif resultado in [

            "Cerca de la meta",

            "Exceso leve"

        ]:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="FFF4E8"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=naranja,

                bold=True

            )


        else:


            hoja_resumen[
                f"F{fila}"
            ].fill = PatternFill(

                "solid",

                fgColor="FFF3F3"

            )


            hoja_resumen[
                f"F{fila}"
            ].font = Font(

                color=rojo,

                bold=True

            )


    # Congelar tabla

    hoja_resumen.freeze_panes = "A12"


    # Filtro

    hoja_resumen.auto_filter.ref = (

        f"A11:F{hoja_resumen.max_row}"

    )


    # Anchos de columnas

    hoja_resumen.column_dimensions["A"].width = 22

    hoja_resumen.column_dimensions["B"].width = 22

    hoja_resumen.column_dimensions["C"].width = 18

    hoja_resumen.column_dimensions["D"].width = 18

    hoja_resumen.column_dimensions["E"].width = 16

    hoja_resumen.column_dimensions["F"].width = 25


    # -----------------------------------------------------
    # 17. GUARDAR EXCEL FINAL
    # -----------------------------------------------------

    libro.save(
        nombre_excel
    )


    # -----------------------------------------------------
    # 18. MOSTRAR CONFIRMACIÓN
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        margin-top:10px;
        margin-bottom:18px;
    ">

        <div style="
            font-size:28px;
            font-weight:bold;
            color:#1F4E79;
        ">
            💾 Tu mes financiero está listo
        </div>


        <div style="
            font-size:15px;
            color:#303030;
            margin-top:8px;
        ">

            👤 <b>{estudiante}</b>

            &nbsp;&nbsp; | &nbsp;&nbsp;

            📅 <b>{mes}</b>

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 19. MOSTRAR DATOS QUE SE GUARDARON
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        display:flex;
        gap:12px;
        flex-wrap:wrap;
        margin-bottom:20px;
    ">


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #2a78d6;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                🧾 Gastos registrados
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1F4E79;
            ">
                {len(historial_exportar)}
            </div>

        </div>


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #eb6834;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💸 Gasto final
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#eb6834;
            ">
                ${total_gastos_exportar:,.0f}
            </div>

        </div>


        <div style="
            flex:1;
            min-width:170px;
            background:#f5f7fa;
            border-top:5px solid #1baf7a;
            border-radius:8px;
            padding:14px;
        ">

            <div style="
                font-size:12px;
                color:#52514e;
            ">
                💚 Ahorro real
            </div>

            <div style="
                font-size:25px;
                font-weight:bold;
                color:#1baf7a;
            ">
                ${ahorro_exportar:,.0f}
            </div>

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 20. MOSTRAR QUÉ CONTIENE EL ARCHIVO
    # -----------------------------------------------------

    display(HTML(f"""
    <div style="
        font-family:Arial,sans-serif;
        background:#eef8f4;
        border-left:6px solid #1baf7a;
        border-radius:8px;
        padding:18px;
        margin-top:15px;
        color:#184f3d;
    ">

        <div style="
            font-size:21px;
            font-weight:bold;
            margin-bottom:8px;
        ">
            ✅ Archivo creado correctamente
        </div>

        <div style="
            font-size:14px;
            line-height:1.8;
        ">

            Tu archivo se llama:

            <br>

            📊 <b>{nombre_excel}</b>

            <br><br>

            <b>Hoja 1 — Historial de gastos</b>

            <br>

            Contiene todos los movimientos que
            registraste en la Celda 3.

            <br><br>

            <b>Hoja 2 — Resumen financiero</b>

            <br>

            Contiene el resultado final utilizado
            en tu cierre del mes:

            <br>

            • Mensualidad

            <br>

            • Gastos finales

            <br>

            • Ahorro real

            <br>

            • Déficit

            <br>

            • Plan vs. Realidad

            <br>

            • Resultado de cada categoría

        </div>

    </div>
    """))


    # -----------------------------------------------------
    # 21. SI EL CIERRE FUE MANUAL, EXPLICAR LA DIFERENCIA
    # -----------------------------------------------------

    if (
        modo_cierre
        == "Ingresar valores finales manualmente"
    ):


        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:16px;
            margin-top:15px;
            color:#765700;
        ">

            <div style="
                font-size:18px;
                font-weight:bold;
                margin-bottom:6px;
            ">
                ℹ️ Importante sobre tu cierre manual
            </div>

            La hoja <b>Historial de gastos</b>
            muestra únicamente lo que registraste
            durante el mes en la aplicación.

            <br><br>

            La hoja <b>Resumen financiero</b>
            utiliza los valores finales manuales
            que confirmaste en la Celda 5.

            <br><br>

            Por eso ambos valores pueden ser diferentes,
            y eso es correcto.

        </div>
        """))


    # -----------------------------------------------------
    # 22. MOSTRAR ENLACE DE DESCARGA
    # -----------------------------------------------------

    display(HTML("""
    <div style="
        font-family:Arial,sans-serif;
        margin-top:20px;
        font-size:19px;
        font-weight:bold;
        color:#1F4E79;
    ">
        📥 Descarga tu archivo Excel:
    </div>
    """))


    display(

        FileLink(
            nombre_excel
        )

    )


    # -----------------------------------------------------
    # 23. INTENTAR DESCARGA AUTOMÁTICA
    # -----------------------------------------------------

    try:

        colab_files.download(
            nombre_excel
        )


    except:

        display(HTML("""
        <div style="
            font-family:Arial,sans-serif;
            background:#fffbea;
            border-left:6px solid #e0a800;
            border-radius:8px;
            padding:15px;
            margin-top:15px;
            color:#765700;
        ">

            ⚠️ El navegador no inició la descarga automáticamente.

            <br><br>

            Haz clic en el enlace que aparece arriba
            para descargar el archivo manualmente.

        </div>
        """))
```
